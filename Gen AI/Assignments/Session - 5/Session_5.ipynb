Use GitHub Copilot or ChatGPT to help you write a function that takes a list of song names and returns only those that start with the letter 'S', like a Spotify playlist filter.<br><br><em><strong>Constraint:</strong> You must show the prompt you used and the code Copilot or ChatGPT generated.</em>

**My Prompt to ChatGPT:**

Write a Python function that takes a list of song names and returns a new list containing only the songs that start with the letter 'S'.

**ChatGPT's Generated Code:**

```python
def filter_songs_by_s(song_list):
    # Use a list comprehension to filter songs starting with 'S' or 's'
    return [song for song in song_list if song.lower().startswith('s')]

# Example usage generated by ChatGPT
spotify_playlist = ["Shape of You", "Blinding Lights", "Starboy", "Levitating", "Smells Like Teen Spirit"]
s_songs = filter_songs_by_s(spotify_playlist)
print("Filtered Playlist:", s_songs)
```

* 5. List two production challenges companies face when using generative AI models for chatbots in apps like BookMyShow or Myntra, and suggest one solution for each challenge.

----

#### ANS:

**1. Challenge 1: Hallucinations (Making up facts).** An AI chatbot for Myntra might confidently tell a customer that an out-of-stock dress is currently available in their size, or invent a fake return policy, leading to extreme customer frustration and lost trust.

* **Solution:** Implement Retrieval-Augmented Generation (RAG). Instead of letting the AI guess, force the chatbot to only generate answers based on a strict, real-time database of the actual store inventory and official policy documents.


**2. Challenge 2: High Latency & Compute Costs.** Running massive GenAI models for thousands of simultaneous BookMyShow customers asking simple questions (e.g., 'What time is the movie?') is incredibly expensive and slow compared to traditional bots.

* **Solution:** Use Hybrid Routing. Put a cheap, fast traditional ML intent-classifier in front of the GenAI. If the question is simple, route it to a fast traditional database lookup. Only route complex, highly conversational queries to the expensive GenAI model.
