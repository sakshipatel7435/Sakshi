#### Comparison with traditional AI/ML

* 1. Create a table comparing traditional AI/ML and Gen AI across at least five aspects such as data requirements, output type, human interaction, adaptability, and example use cases. Use real-world apps you know (like Instagram, Zomato, or Spotify) as examples for each.

----

#### ANS:

| Aspect | Traditional AI/ML | Generative AI (GenAI) |
| :--- | :--- | :--- |
| **Output Type** | Predictive / Fixed (Numbers, Categories, Rankings). | Creative / New (Text, Images, Audio, Code). |
| **Data Requirements** | Requires massive amounts of structured, labeled data (e.g., thousands of photos labeled 'cat'). | Pre-trained on massive unstructured datasets (like the entire internet), then fine-tuned. |
| **Human Interaction** | Passive. Users receive a result (e.g., a credit score or a recommended song). | Conversational / Interactive. Users provide 'prompts' and iterate with the AI. |
| **Adaptability** | Rigid. A model trained to detect spam cannot suddenly write a poem. | Highly flexible. An LLM can write code, translate languages, and write poems simultaneously. |
| **Example Use Case** | **Spotify Recommendations:** Predicting what existing song you want to hear next. | **Instagram AI Avatars:** Generating completely new, stylized artwork of your face. |

* 2. Given a scenario where Zomato recommends restaurants based on your previous orders, explain whether this uses traditional AI/ML or Gen AI, and justify your answer in 3-4 sentences.

----

#### ANS:

Zomato recommending restaurants based on your previous orders is a classic example of Traditional AI/ML. It uses recommendation algorithms (like collaborative filtering) to find patterns in historical data and match you with existing restaurants that similar users liked. It is not creating a brand-new, fictional restaurant or generating novel food items from scratch; it is simply predicting a preference and retrieving fixed, existing data.

* 4. Imagine you are building a chatbot for Flipkart customer support. List two ways a traditional AI/ML approach would handle user queries, and two ways a Gen AI approach would differ in handling the same queries.<br><br><em><strong>Hint:</strong> Focus on how responses are generated and the flexibility of the answers.</em>

----

#### ANS:

**Traditional AI/ML Approach:**

1. Uses rigid decision trees and predefined keyword matching (e.g., if the user types 'refund', it instantly replies with a hardcoded template: 'Please wait 5-7 days').
2. Fails completely if the user asks a complex or misspelled question outside of its pre-programmed rules, resulting in 'I didn't understand that, please select an option.'


**Gen AI Approach:**

1. Dynamically generates natural, highly flexible responses by understanding the true context of the entire conversation, even if the user makes typos or explains their issue in a confusing way.
2. Can instantly synthesize a unique, personalized solution (e.g., summarizing the exact warranty policy for the specific blender the user bought) instead of forcing the user to read a generic FAQ page.
