#### ANS:
| Point | Reinforcement Learning | Supervised Learning | Unsupervised Learning |
| :--- | :--- | :--- | :--- |
| **Learning method** | Learns by taking actions and receiving rewards or penalties. | Learns from labeled data where the correct answer is given. | Learns patterns from data without labeled answers. |

#### ANS: Zomato Restaurant Recommendations (Reinforcement Learning)
* **Agent:** The Zomato recommendation engine algorithm.

* **Environment:** The app interface, user context (current location, time of day, weather, past order history), and the catalog of available restaurants.

* **Actions:** Deciding which restaurant or dish card to display at the top of the user's home screen or search results.

* **Rewards:**

    * **Positive Reward:** The user clicks the recommendation, places an order, and gives a 4 or 5-star rating.

    * **Negative Reward:** The user scrolls past the recommendation, ignores it, or closes the app without ordering.

* 3. Imagine you are designing a simple RL agent for a cricket game app (like IPL fantasy league). Describe what could be the agent, environment, possible actions, and how rewards would be assigned for the agent's choices.

#### ANS: IPL Fantasy League RL Agent Design
* **Agent:** The fantasy squad selection algorithm.

* **Environment:** Upcoming match details (pitch conditions, weather forecast, stadium dimensions, opposition bowler/batter matchups), and player historical statistics/recent form.

* **Actions:** Selecting a playing XI and assigning the Captain (2x points) and Vice-Captain (1.5x points) from the pool of available players.

* **Rewards:**

    * **Positive Reward:** Total fantasy points accumulated by the chosen team during the actual live match (runs scored, wickets taken, catches, strike-rate bonuses).

    * **Negative Reward:** Points lost for ducks, negative economy rates, or choosing a player who ends up sitting out of the match (0 points).

* 4. Think of a feature in any app you use (like YouTube autoplay, Flipkart product suggestions, or Spotify playlist generation) that could be improved using Reinforcement Learning. Briefly describe how RL could make the feature smarter, and what the reward signal might be.

#### ANS: YouTube Autoplay Feature Improvement
* **Current Limitation:** Traditional autoplay often loops into a rigid rabbit hole of the exact same creator or overly similar videos, leading to user fatigue.

* **How RL Makes it Smarter:** An RL agent can optimize for long-term session satisfaction rather than just immediate clicks. It learns to balance exploitation (showing similar videos) with exploration (introducing a fresh, adjacent topic) to keep the user engaged across a longer viewing session.

* **Reward Signal:**

    * **Positive:** High watch time, video completion rate, and the user continuing to browse the next recommended video.

    * **Negative:** The user manually stops the video, hits the back button within the first 10 seconds, or closes the app entirely.
