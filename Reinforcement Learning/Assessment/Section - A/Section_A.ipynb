### Section A — Concept Application

### 1. SCENARIO
You are a data scientist at a food delivery company building an AI system that must
automatically improve its delivery dispatching decisions over time, without being given
any labelled examples of correct or incorrect dispatch choices.

* Question: Explain why Reinforcement Learning is more suitable than traditional programming
for this dispatch system. What key characteristic of RL allows it to improve its decisions without
relying on labelled training data?

### ANS:
Reinforcement Learning is perfect here because dispatching is highly dynamic and unpredictable (traffic, weather, delays). Traditional programming would require thousands of hardcoded IF/ELSE rules, which is impossible to scale.

The key characteristic of RL is Trial-and-Error through Rewards. It doesn't need humans to label 'correct' choices; it simply tries different dispatch routes and learns by maximizing a numeric reward signal (like getting +10 points for a fast delivery).

### 2. SCENARIO
You are building a restaurant recommendation engine and your manager suggests
training a supervised learning model on historical order data. A colleague argues that a
Reinforcement Learning agent would produce better results over time.

* Question: Compare how a supervised learning model and a Reinforcement Learning agent
each learn to make recommendations. What fundamental difference in their learning
mechanism would justify choosing RL for a continuously improving recommendation engine?

### ANS:
A supervised learning model trains on static historical order data to map features to a fixed target label (e.g., predicting if a user will buy a specific dish based on past habits). Conversely, an RL agent treats recommendation as a sequential decision-making process, exploring new choices and exploiting known preferences. The fundamental difference is that RL learns dynamically from feedback loops and delayed consequences, allowing it to adapt in real-time when user preferences shift, whereas supervised models remain bounded by the fixed distribution of past training data.

### 3. SCENARIO
You are evaluating whether to use unsupervised learning (clustering) or Reinforcement
Learning to optimise the delivery zones assigned to riders on your platform. Both
approaches work without labelled data, yet your technical lead says they are
fundamentally different.

* Question: What distinguishes a Reinforcement Learning system from an unsupervised learning
system when the goal is to continuously improve delivery zone assignments based on real
delivery outcomes?

### ANS:
Unsupervised learning (such as clustering) groups data based on hidden patterns or structural similarities without any feedback loop or notion of "good" versus "bad" outcomes. A Reinforcement Learning system, however, actively takes actions, observes the resulting environmental changes (e.g., delivery times and rider efficiency), and receives explicit rewards or penalties. This feedback loop allows the RL system to explicitly optimize its policy for better performance over time.

### 4. SCENARIO
You are designing an RL agent for your food delivery platform that must decide which
rider to assign to each incoming order. Your team needs to define the core components
of the RL system before implementation begins.

* Question: Identify the agent, environment, set of possible actions, and reward signal for this
delivery dispatch RL system. Justify why your chosen reward signal directly encourages efficient
delivery performance rather than unintended shortcut behaviour.


### ANS:

* Agent: The AI dispatch system.
* Environment: The food delivery platform (including riders, real-time traffic, incoming orders, and customers).
* Actions: Assigning specific riders to specific orders, or deferring/rejecting assignments.
* Reward Signal: $+10$ for on-time delivery, plus the customer rating (1–5), minus penalties for late deliveries or high travel distance.
* Justification: Incorporating delivery punctuality and customer ratings penalizes reckless or dangerous "shortcut" behaviors that result in poor service, damaged items, or customer complaints, thereby steering the agent toward safe, efficient delivery performance.

### 5. SCENARIO
You are presenting Reinforcement Learning to your company's product team, who are
familiar with how game-playing AI (such as chess engines) works. They want to
understand how the same principle applies to food delivery.

* Question: Using game AI as a reference point, explain the core principle of how an RL agent
learns through interaction with its environment. In what ways does the delivery dispatch scenario
follow this same learning principle even though it is not a game?

### ANS:
In game AI (like chess), the agent observes the board state, makes a move, transitions to a new state, and eventually receives a win/loss reward at the end. The core principle is mapping states to actions to maximize cumulative future rewards. Similarly, the delivery dispatch scenario follows this exact Markov Decision Process (MDP) framework: the current state consists of active orders and rider locations; the dispatch decision acts as the move; and the delivery outcome (time and rating) acts as the reward signal.

### 6. SCENARIO
You are advising your company on adopting an RL-based system to personalise meal
suggestions for each customer on your platform, similar to how streaming services use
RL to personalise content recommendations.

* Question: Describe how Reinforcement Learning has been applied to recommendation systems
in industry. What advantage does an RL-based recommendation system offer over a model
trained once on historical data, particularly for customers whose food preferences change over
time?


### ANS:
In industry, RL recommendation systems treat item presentation as a multi-step dialogue with the user, maximizing long-term engagement rather than just immediate clicks. The primary advantage over static historical models is adaptability. Static models become stale when user tastes change or seasonal trends shift, whereas an RL agent continuously explores new recommendations and updates its policy based on immediate user feedback.
