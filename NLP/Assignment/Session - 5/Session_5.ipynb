```python
from transformers import pipeline
```

```python
sentiment_pipeline = pipeline('sentiment-analysis')
```

```python
reviews = [
    'The delivery was incredibly fast and the food was hot!',
    'Worst biryani I have ever had, completely tasteless.',
    'Decent portion size but a bit overpriced for what you get.'
]
```

```python
for review in reviews:
    result = sentiment_pipeline(review)[0]
    print(f"Review: '{review}'\nPredicted Sentiment: {result['label']} (Score: {result['score']})\n")
```

Use ChatGPT or Copilot to generate a simple Python function that demonstrates the attention mechanism with three short sentences as input, and explain in your own words (2-3 lines) what the attention scores mean in your output.<br><br><em><strong>Hint:</strong> Ask the AI tool to 'show a basic attention calculation example in Python.'</em>

```python
import numpy as np

def simple_attention(query, keys, values):
    scores = np.dot(query, keys.T)
    weights = np.exp(scores) / np.sum(np.exp(scores), axis=-1, keepdims=True)
    output = np.dot(weights, values)
    return output, weights
```

```python
sentences = ['I love food', 'Food is good', 'I am hungry']
# Dummy embeddings for the 3 sentences (size=4)
embeddings = np.array([
    [0.1, 0.2, 0.3, 0.4],
    [0.2, 0.3, 0.4, 0.5],
    [0.9, 0.8, 0.7, 0.6]
])
```

```python
query = embeddings 
keys = embeddings
values = embeddings

output, attention_weights = simple_attention(query, keys, values)

print('Attention Weights Matrix:\n', np.round(attention_weights, 4))
```
