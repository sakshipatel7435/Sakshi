* 3. Given an array of WhatsApp chat messages, count and display the frequency of each word that appears across all messages.<br><br><em><strong>Hint:</strong> Use a dictionary to store word counts and ignore common stopwords like 'the', 'is', 'at'.</em>

```python
messages = ['The party is at my house', 'Is the house party tonight', 'See you at the party']
stopwords = {'the', 'is', 'at', 'you', 'my'}
word_counts = {}

for msg in messages:
    words = msg.lower().split()
    for word in words:
        clean_word = word.strip(string.punctuation)
        if clean_word and clean_word not in stopwords:
            word_counts[clean_word] = word_counts.get(clean_word, 0) + 1

for word, count in word_counts.items():
    print(f"'{word}': {count}")
```

* 4. Use scikit-learn's TfidfVectorizer to transform a list of Flipkart product reviews into TF-IDF vectors, then print the TF-IDF value for the word 'quality' in each review.<br><br><em><strong>Constraint:</strong> Use at least 3 reviews and show the output for each.</em>

```python
flipkart_reviews = [
    'The quality of this camera is amazing.',
    'Fast delivery but average quality.',
    'I did not like the product at all.'
]
```

```python
tfidf_vec = TfidfVectorizer()
tfidf_matrix = tfidf_vec.fit_transform(flipkart_reviews)
vocab = tfidf_vec.get_feature_names_out()
```

```python
tfidf_matrix.toarray()
```

```python
vocab
```

```python
if 'quality' in vocab:
    quality_index = list(vocab).index('quality')
    for i, review in enumerate(flipkart_reviews):
        score = tfidf_matrix.toarray()[i][quality_index]
        print(f"Review {i+1} TF-IDF for 'quality': {score}")
else:
    print("'quality' is not in the vocabulary.")
```
