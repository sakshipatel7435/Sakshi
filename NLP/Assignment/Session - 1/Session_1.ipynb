* 4. Build a simple sentiment checker in Python: ask the user to enter a review about a movie (like on BookMyShow), then check if the review contains the word 'good' or 'bad' and print 'Positive', 'Negative', or 'Neutral'.<br><br><em><strong>Constraint:</strong> Only use basic Python (no external libraries).</em>

```python
review = input("Enter a movie review (e.g., for BookMyShow): ").lower()

if "good" in review and "bad" in review:
    print("Neutral") 
elif "good" in review:
    print("Positive")
elif "bad" in review:
    print("Negative")
else:
    print("Neutral")
```

* 5. Imagine you are designing a spam filter for an email app. Write a Python function is_spam(email_text) that returns True if the email_text contains the word 'win' or 'free', otherwise returns False. Test your function with two examples.

```python
def is_spam(email_text):
    text_lower = email_text.lower()
    
    if "win" in text_lower or "free" in text_lower:
        return True
    return False

example_1 = "Congratulations! You win a free vacation package now!"
example_2 = "Hey, are we still meeting for lunch today at 1 PM?"

print(f"Email 1 Is Spam: {is_spam(example_1)}")
print(f"Email 2 Is Spam: {is_spam(example_2)}")
```
