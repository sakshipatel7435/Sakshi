* 2. Create a function to remove all punctuation marks from a Flipkart product review string and return the cleaned text.<br><br><em><strong>Hint:</strong> Use the string.punctuation constant from the Python string module.</em>

```python
import string
```

```python
def remove_punctuation(review):
    clean_text = ""
    for char in review:
        if char not in string.punctuation:
            clean_text = clean_text + char
    return clean_text

review_text = "Wow! This product is amazing, I love it."
print(remove_punctuation(review_text))
```

* 4. Download a list of English stopwords using NLTK and write a function that removes all stopwords from a WhatsApp message string.

```python
import nltk
```

```python
nltk.download('stopwords', quiet=True)
```

```python
from nltk.corpus import stopwords
```

```python
def remove_stopwords(message):
    stop_words = set(stopwords.words('english'))
    words = message.split()
    clean_words = []
    for word in words:
        if word.lower() not in stop_words:
            clean_words.append(word)
    return " ".join(clean_words)

whatsapp_msg = "I am going to the store to buy some milk and bread"
print(remove_stopwords(whatsapp_msg))
```

* 5. Use NLTK’s PorterStemmer to stem each word in a list of trending hashtags from Twitter, and print both the original and stemmed versions side by side.<br><br><em><strong>Hint:</strong> Example hashtags: ['#playing', '#dancing', '#eating']</em>

```python
from nltk.stem import PorterStemmer
```

```python
stemmer = PorterStemmer()
hashtags = ['#playing', '#dancing', '#eating', '#running', '#coding']

print(f"{'Original Hashtag':<20} | {'Stemmed Word':<15}")
print("-" * 38)

for hashtag in hashtags:
    word = hashtag.replace("#", "")
    stemmed_word = stemmer.stem(word)
    print(f"{hashtag:<20} | #{stemmed_word:<15}")
```
