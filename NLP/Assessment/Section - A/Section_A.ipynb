### Section A — Concept Application

### 1. SCENARIO
You are building an automated customer support chatbot for a food delivery platform
that handles queries like 'Where is my order?', 'I want to cancel my order', and 'My food
arrived cold and late'

* Question: Identify two NLP techniques the chatbot must use to correctly understand and route
these customer messages. Explain why each technique is necessary and describe what would
happen to chatbot accuracy if either technique were removed.

### ANS:

1. **Intent Classification (Intent Recognition):** Identifies what the user wants to accomplish (e.g., tracking an order, canceling, or filing a quality complaint). It routes the request to the appropriate logic path or support workflow. Without it, the bot cannot map varied user inputs to specific system actions, leading to misrouted or generic requests.

2. **Named Entity Recognition (NER):** Extracts key parameters/entities from text (e.g., order ID, food item names, delivery condition attributes like "cold" or "late"). Without it, the bot might know the customer wants to cancel, but won't know which order or item to target without additional tedious prompts.

### 2. SCENARIO
You are processing 50,000 raw customer reviews scraped from a food delivery app. The
reviews contain mixed letter cases, special characters, emojis, numbers, and common
filler words such as 'the', 'is', and 'a'.

* Question: Describe the text preprocessing pipeline you would apply to this dataset before
feeding it to any NLP model. Justify the order in which you would apply each step, and explain
specifically why skipping stopword removal could hurt downstream model performance.

### ANS:

Preprocessing Pipeline Order:
1. **Noise Removal & Text Cleaning:** Remove HTML tags, emojis, special characters, numbers, and extra spaces.
2. **Lowercasing:** Convert all text to lower case to ensure standardizing ("Cold" vs "cold").
3. **Tokenization:** Split text into individual words/tokens.
4. **Stopword Removal:** Remove low-information words ("the", "is", "a").   
5. **Stemming / Lemmatization:** Reduce words to root form ("delivering" $\rightarrow$ "deliver").

* **Justification:** Noise removal and lowercasing must happen first so that tokenization and stopword lookup match correctly against clean, lowercased dictionaries.
* **Impact of skipping Stopword Removal:** Non-informative filler words will dominate feature counts and sparse matrices, diluting the TF-IDF weight or probabilistic importance of domain-critical words (like "cold", "late", "missing").  

### 3. SCENARIO
You are building a food item search feature that must rank menu descriptions by
relevance to a customer's typed query, for example 'spicy chicken without onion'.


* Question: Compare Bag of Words and TF-IDF as text representation methods for this ranking
task. Which would you choose and why? Identify one limitation that your chosen method still has
even after it is applied.

### ANS:

**Comparison & Choice:** TF-IDF is superior to Bag of Words (BoW) because BoW treats all word frequencies equally. TF-IDF penalizes common words (like "food" or "good") across all menu descriptions while highlighting distinctive query terms (like "spicy" or "onion").

**Limitation:** Neither method accounts for word context, order, or semantics (synonyms/negation handling). For example, TF-IDF cannot naturally understand that "without onion" requires excluding onions, or that "spicy" and "hot" share semantic similarity.

### 4. SCENARIO
You are training a Naive Bayes classifier to automatically label incoming food delivery
support tickets as 'Delivery Issue', 'Food Quality Issue', or 'App Issue' using a dataset of
10,000 historical labelled tickets.


* Question: Explain the core probabilistic assumption that Naive Bayes makes about its input
features. Describe a realistic food delivery example where this assumption is violated, and
explain how that violation would affect the model's classification accuracy.


### ANS:

**Core Assumption:** Naive Bayes assumes conditional independence among features given the class label.

**Violation Example:** In food complaints, the words "arrived" and "cold" or "extremely" and "late" often co-occur dependently.

**Effect on Accuracy:** The model will overcount evidence by treating correlated words as independent evidence signals, leading to overconfident or biased probability predictions.

### 5. SCENARIO
You are comparing a Logistic Regression model and a Naive Bayes model for classifying
food delivery reviews as Positive, Neutral, or Negative. Your dataset is heavily
imbalanced: 80% of reviews are Positive, 15% Neutral, and 5% Negative.


* Question: Describe two evaluation metrics you would prioritise over accuracy when comparing
these models on this imbalanced dataset. Explain one technique you would apply to address the
class imbalance before training.

### ANS:

**Metrics:** Macro $F_1$-Score (or Precision-Recall AUC) and Balanced Accuracy / Recall per class. Accuracy gives a misleading impression on 80/15/5 imbalanced datasets because predicting the majority class exclusively still yields 80% accuracy.

**Class Imbalance Technique:** SMOTE (Synthetic Minority Over-sampling Technique) or class-weighted loss adjustment (class_weight='balanced') during model training.

### 6. SCENARIO
You are integrating a Hugging Face transformer pipeline into your food delivery app to
automatically generate polite responses to customer reviews in real time, at a rate of
several hundred requests per minute.

* Question: Explain what the attention mechanism in transformers enables that traditional
RNN-based models could not do effectively with long text sequences. Describe one practical
challenge of running a transformer model in a high-traffic production environment and one
common way to mitigate it.

### ANS:

**Attention Mechanism Benefit:** It allows parallel processing of entire sequences and direct modeling of dependencies between distant tokens without vanishing gradients, unlike sequential RNNs.

**Production Challenge:** High latency and high memory footprint when processing hundreds of requests per minute.

**Mitigation:** Model quantization/distillation (e.g., using DistilBERT or ONNX Runtime inference engine).
