### Task 1: Food Review Text Preprocessor

Build a reusable text preprocessing function that cleans raw food delivery customer reviews for downstream NLP tasks.

* Requirement 1 — Accept a raw review string as input and return a cleaned, processed string.
* Requirement 2 — Convert text to lowercase, remove punctuation, and strip extra whitespace.
* Requirement 3 — Tokenise the cleaned text and remove English stopwords using NLTK's stopwords corpus.
* Requirement 4 — Apply stemming using PorterStemmer (or lemmatization using WordNetLemmatizer) to the token list and return the result as a single space-joined string. Test your function on at least three different food delivery review samples.


```python
import string
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
nltk.download('stopwords', quiet=True)
nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)
```

```python
def preprocess_review(review):
    review = review.lower()
    for p in string.punctuation:
        review = review.replace(p, "")
        
    words = review.split()
    stop_words = stopwords.words('english')
    clean_words = []
    
    for w in words:
        if w not in stop_words:
            clean_words.append(w)
    stemmer = PorterStemmer()
    final_words = []
    for w in clean_words:
        final_words.append(stemmer.stem(w))
        
    return " ".join(final_words)
```

```python
sample_reviews = [
    "The food was absolutely AMAZING! Loved it.",
    "Terrible delivery time, my pizza arrived ice cold...",
    "Decent taste, but WAY too overpriced for this quantity."
]

for rev in sample_reviews:
    print(f'Original: {rev}')
    print(f'Processed: {preprocess_review(rev)}\n')
```

### Task 2: TF-IDF Vectoriser for Menu Reviews

Apply TF-IDF vectorisation to a set of food delivery menu item reviews and identify the most informative terms per document.
* Requirement 1 — Create a list of at least 8 short food delivery review strings as your corpus (mix of positive, negative, and neutral tones).
* Requirement 2 — Use sklearn's TfidfVectorizer to transform the corpus into a TF-IDF matrix.
* Requirement 3 — Display the full TF-IDF matrix as a labelled pandas DataFrame with feature names as column headers and review index as row labels.
* Requirement 4 — For each review, print the top 3 words with the highest TF-IDF score and their scores.


```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
menu_corpus = [
    "Spicy paneer tikka with mint chutney",
    "Bland and cold french fries",
    "Best chocolate lava cake ever",
    "Average taste but good portion size",
    "Stale burger buns and smelly cheese",
    "Freshly baked garlic bread with cheese",
    "Very oily and salty noodles",
    "Perfectly cooked medium rare steak"
]
```

```python
tfidf_vec = TfidfVectorizer()
tfidf_matrix = tfidf_vec.fit_transform(menu_corpus)
feature_names = tfidf_vec.get_feature_names_out()

df_tfidf = pd.DataFrame(tfidf_matrix.toarray(), columns=feature_names, index=[f'Review {i+1}' for i in range(len(menu_corpus))])
display(df_tfidf)
```

```python
for i, row in df_tfidf.iterrows():
    top_3 = row.nlargest(3)
    print(f"{i}:")
    for word, score in top_3.items():
        print(f"  {word}: {score:.4f}")
```

### Task 3: Complaint Category Classifier – Naive Bayes



Train a Multinomial Naive Bayes text classifier to automatically categorise food delivery support complaints into predefined issue types.
* Requirement 1 — Create a labelled dataset of at least 24 complaint strings across three classes: 'Delivery', 'Food Quality', and 'App'.
* Requirement 2 — Preprocess each complaint using the function from Task 1, then convert the corpus to a TF-IDF feature matrix.
* Requirement 3 — Split into 80% train and 20% test sets, train a MultinomialNB classifier, and generate predictions on the test set.
* Requirement 4 — Print the full classification report (precision, recall, F1-score per class) and the confusion matrix.


```python
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import classification_report, confusion_matrix
```

```python
complaints = [
    "Order arrived two hours late", "The pizza was completely crushed in the box", "App keeps crashing when I try to pay",
    "Delivery boy was extremely rude", "Food is stale and smells bad", "Payment went through but order not placed",
    "Missing items in my delivery", "The curry was way too salty", "Cannot apply promo code on the checkout screen",
    "Food was delivered to the wrong address", "Found a hair in my soup", "Location tracking is not updating on the map",
    "Delayed delivery again", "Chicken is undercooked", "Screen freezes when selecting restaurant",
    "Delivery took forever", "Ice cream melted completely", "Unable to login to my account",
    "Driver refused to come upstairs", "Vegetables are not fresh", "OTP verification is failing",
    "Never received my order", "Very bad food quality", "App shows network error constantly"
]

labels = [
    "Delivery", "Food Quality", "App",
    "Delivery", "Food Quality", "App",
    "Delivery", "Food Quality", "App",
    "Delivery", "Food Quality", "App",
    "Delivery", "Food Quality", "App",
    "Delivery", "Food Quality", "App",
    "Delivery", "Food Quality", "App",
    "Delivery", "Food Quality", "App"
]
```

```python
processed_complaints = [preprocess_review(c) for c in complaints]

vec_complaints = TfidfVectorizer()
X_complaints = vec_complaints.fit_transform(processed_complaints)

X_train, X_test, y_train, y_test = train_test_split(X_complaints, labels, test_size=0.2, random_state=42)
```

```python
nb_classifier = MultinomialNB()
nb_classifier.fit(X_train, y_train)
y_pred = nb_classifier.predict(X_test)

print("Classification Report:\n", classification_report(y_test, y_pred, zero_division=0))
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
```

### Task 4: Sentiment Analysis Pipeline – Model Comparison

Build a complete end-to-end NLP pipeline that preprocesses food delivery reviews, vectorises them, and compares Naive Bayes against Logistic Regression on a binary sentiment task.
* Create a labelled dataset of at least 30 food delivery reviews with 'Positive' and 'Negative'
sentiment labels.
* Build two separate sklearn Pipeline objects — one combining TfidfVectorizer with
MultinomialNB and one combining TfidfVectorizer with LogisticRegression.
* Train both pipelines and evaluate each on a held-out test set; collect accuracy, precision,
recall, and F1-score.
* Print a side-by-side comparison table of both models' metrics and include a one-line
comment in code stating which model you would deploy and your reason.

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
```

```python
sentiment_reviews = [
    "Delicious food", "Terrible taste", "Loved the delivery speed", "Worst experience",
    "Highly recommend", "Never ordering again", "Fresh and hot", "Stale food",
    "Amazing quality", "Very bad", "Best pizza ever", "Waste of money",
    "Good portion size", "Too salty", "Will order again", "Disgusting",
    "Perfectly cooked", "Undercooked meat", "Value for money", "Overpriced",
    "Great packaging", "Box was torn", "Friendly rider", "Rude delivery guy",
    "Authentic taste", "Tasteless", "Super fast", "Took two hours",
    "Five stars", "Zero stars"
]

sentiment_labels = [
    "Positive", "Negative", "Positive", "Negative",
    "Positive", "Negative", "Positive", "Negative",
    "Positive", "Negative", "Positive", "Negative",
    "Positive", "Negative", "Positive", "Negative",
    "Positive", "Negative", "Positive", "Negative",
    "Positive", "Negative", "Positive", "Negative",
    "Positive", "Negative", "Positive", "Negative",
    "Positive", "Negative"
]
```

```python
X_train_sent, X_test_sent, y_train_sent, y_test_sent = train_test_split(
    sentiment_reviews, sentiment_labels, test_size=0.3, random_state=42
)
```

```python
pipe_nb = Pipeline([
    ('tfidf', TfidfVectorizer(preprocessor=preprocess_review)),
    ('nb', MultinomialNB())
])

pipe_lr = Pipeline([
    ('tfidf', TfidfVectorizer(preprocessor=preprocess_review)),
    ('lr', LogisticRegression())
])

pipe_nb.fit(X_train_sent, y_train_sent)
pipe_lr.fit(X_train_sent, y_train_sent)

pred_nb = pipe_nb.predict(X_test_sent)
pred_lr = pipe_lr.predict(X_test_sent)
```

```python

def get_metrics(y_true, y_pred):
    return {
        'Accuracy': accuracy_score(y_true, y_pred),
        'Precision': precision_score(y_true, y_pred, pos_label='Positive'),
        'Recall': recall_score(y_true, y_pred, pos_label='Positive'),
        'F1-Score': f1_score(y_true, y_pred, pos_label='Positive')
    }

metrics_nb = get_metrics(y_test_sent, pred_nb)
metrics_lr = get_metrics(y_test_sent, pred_lr)

df_metrics = pd.DataFrame([metrics_nb, metrics_lr], index=['Naive Bayes', 'Logistic Regression'])
display(df_metrics)

print("# I would deploy Logistic Regression because it typically handles sparse TF-IDF vectors and binary classification decision boundaries much better than Naive Bayes on smaller datasets.")
```
