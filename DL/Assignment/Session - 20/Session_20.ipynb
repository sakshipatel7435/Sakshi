* 1. Create a 6x6 NumPy array with random integers (0-255) to simulate a grayscale image, then apply a 2x2 max pooling operation with a stride of 2 and print the resulting reduced array.

```python
import numpy as np
```

```python
np.random.seed(42)
```

```python
image_1 = np.random.randint(0, 255,size=(6, 6))
```

```python
pool_size = 2
stride = 2
```

```python
out_h = image_1.shape[0] // stride
out_w = image_1.shape[1] // stride
```

```python
max_pooled_1 = np.zeros((out_h, out_w), dtype=int)
```

```python
for i in range(out_h):
    for j in range(out_w):
        r = i * stride
        c = j * stride
        region = image_1[r:r+pool_size, c:c+pool_size]
        max_pooled_1[i, j] = np.max(region)
```

```python
print("* Simulated 6X6 GrayScale Image:")
print(image_1)
print("\n* Reduced 3X3 Max Pooled Output:")
print(max_pooled_1)
```

* 2. Write a Python function average_pooling_2x2(image) that takes a 2D NumPy array and returns a new array after applying 2x2 average pooling with stride 2.

```python
def average_pooling_2x2(image):
    pool_size = 2
    stride = 2

    in_h, in_w = image.shape
    out_h = in_h // stride
    out_w = in_w // stride

    pooled_array = np.zeros((out_h, out_w), dtype=np.float32)

    for i in range(out_h):
        for j in range(out_w):
            r = i * stride
            c = j * stride
            
            region = image[r:r+pool_size, c:c+pool_size]
            pooled_array[i, j] = np.mean(region)

    return pooled_array
```

```python
sample_matrix = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [12, 24, 36, 48],
    [60, 72, 84, 96]
])
```

```python
print("* Original Array:")
print(sample_matrix)
print("\n* 2X2 Average Pooled Array:")
print(average_pooling_2x2(sample_matrix))
```

* 3. Given a 4x4 feature map array, manually perform both max pooling and average pooling with a 2x2 window and stride 2, and compare the results side by side.<br><br><em><strong>Hint:</strong> Show both resulting arrays and explain in one line which pooling method keeps more details from the original image.</em>

```python
feature_map = np.array([
    [12, 20,  0,  8],
    [ 4, 32,  2, 14],
    [10, 10, 80, 20],
    [ 2,  6, 40, 10]
])
```

```python
max_pool = np.zeros((2, 2))
avg_pool = np.zeros((2, 2))
```

```python
for i in range(2):
    for j in range(2):
        region = feature_map[i*2:i*2+2, j*2:j*2+2]
        max_pool[i, j] = np.max(region)
        avg_pool[i, j] = np.mean(region)
```

```python
print("* 4X4 Feature Map:")
print(feature_map)
print("\n* Side-by-Side Comparison")
print("Max Pooled (2X2):")
print(max_pool)
print("\nAverage Pooled (2X2):")
print(avg_pool)
```

**Key Difference:**

Average Pooling keeps more overall details and structural background context from the original image by smoothing pixel values, whereas Max Pooling aggressively discards low values to preserve only the most prominent, dominant features (like strong edges).

* 4. Combine a single 3x3 convolution layer (use any simple filter, e.g., edge detection) followed by a 2x2 max pooling layer on a 5x5 input array. Show the output after each step.

```python
input_4 = np.array([
    [1, 2, 3, 0, 1],
    [0, 1, 2, 4, 1],
    [3, 1, 0, 2, 0],
    [1, 0, 1, 3, 2],
    [2, 1, 0, 1, 1]
])

edge_filter = np.array([
    [ 1,  0, -1],
    [ 1,  0, -1],
    [ 1,  0, -1]
])
```

```python
conv_out = np.zeros((3, 3))
for i in range(3):
    for j in range(3):
        region = input_4[i:i+3, j:j+3]
        conv_out[i, j] = np.sum(region * edge_filter)
```

```python
pool_out = np.zeros((2, 2))
for i in range(2):
    for j in range(2):
        region = conv_out[i:i+2, j:j+2]
        pool_out[i, j] = np.max(region)
```

```python
print("1. Original 5x5 Input:")
print(input_4)
print("\n2. After 3x3 Convolution (3x3 Feature Map):")
print(conv_out)
print("\n3. After 2x2 Max Pooling (2x2 Feature Map):")
print(pool_out)
```

* 5. Explain in your own words, using an example from Instagram or Zomato, how pooling layers help reduce computation and improve feature extraction in deep learning models.

**Real-World Concept — Pooling in Instagram/Zomato**

**1. Reducing Computation**

When you upload a high-resolution food photo to Zomato or a selfie to Instagram, the image can contain millions of pixel inputs. A pooling layer downsizes the image representations (e.g., halving the height and width cuts total pixel memory by 75%), drastically reducing the matrix parameters that downstream neural network layers need to process. This enables fast real-time image processing on mobile devices.

**2. Improving Feature Extraction**

If a CNN detects a food dish (e.g., a burger) on Zomato, Max Pooling extracts only the highest activation values from local pixel regions. It focuses on sharp, defining characteristics—such as the curve of the burger bun or the texture of sesame seeds—while ignoring minor background noise or lighting shifts. This ensures that even if the dish moves slightly or has minor background variations, the model reliably identifies the key item.
