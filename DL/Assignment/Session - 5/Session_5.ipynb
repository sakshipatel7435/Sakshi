* 1. Implement the Leaky ReLU activation function in Python as a function leaky_relu(x, alpha=0.01), and test it on an array of values ranging from -5 to 5.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def leaky_relu(x, alpha=0.01):
    return np.maximum(alpha * x, x)

x_vals = np.linspace(-5, 5, 11)
output = leaky_relu(x_vals)
```

```python
print("Inputs: ", x_vals)
print("Outputs:", output)
```

* 3. Given a list of predicted scores for 4 classes: [2.0, 1.0, 0.1, 0.5], implement the softmax function in Python and print the resulting probabilities rounded to 3 decimal places.

```python
def softmax(x):
    exp_x = np.exp(x - np.max(x))       # Subtract max(x) for numerical stability
    return exp_x / np.sum(exp_x)

scores = [2.0, 1.0, 0.1, 0.5]
probabilities = np.round(softmax(scores), 3)
```

```python
print("Predicted Scores:", scores)
print("Probabilities:  ", probabilities.tolist())
```

* 4. Write a function hard_sigmoid(x) that implements the hard sigmoid activation (output between 0 and 1), and test it on inputs: -3, -1, 0, 1, 3.<br><br><em><strong>Hint:</strong> Use the formula: hard_sigmoid(x) = max(0, min(1, 0.2*x + 0.5)).</em>

```python
def hard_sigmoid(x):
    x = np.array(x, dtype=float)
    return np.clip(0.2 * x + 0.5, 0, 1)

inputs = [-3, -1, 0, 1, 3]
outputs = hard_sigmoid(inputs)
```

```python
for inp, out in zip(inputs, outputs):
    print(f"hard_sigmoid({inp:2d}) = {out:.2f}")
```

For each layer below, choose and justify the best activation function: (a) hidden dense layer, (b) output layer with 5 classes, (c) output layer with binary classification.<br><br><em><strong>Constraint:</strong> Write 1-2 lines of reasoning for each choice.</em>

#### (a) Hidden Dense Layer: ReLU (or Leaky ReLU/Swish).

Reason: ReLU avoids the vanishing gradient problem in deep networks and allows fast, computationally efficient feature training.

#### (b) Output Layer with 5 Classes: Softmax.

Reason: Softmax converts unnormalized logits into a mutually exclusive probability distribution across all 5 classes that sums to 1.

#### (c) Output Layer with Binary Classification: Sigmoid.

Reason: Sigmoid squeezes the single output score into a range between 0 and 1, representing the exact probability of belonging to the positive class.
