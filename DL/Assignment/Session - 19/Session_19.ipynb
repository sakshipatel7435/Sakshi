* 1. Create a 5x5 NumPy matrix with random integers from 0 to 9, then define a 3x3 filter (kernel) and manually perform a single convolution operation (no padding, stride 1) to produce the activation map.

```python
import numpy as np
```

```python
np.random.seed(42)
```

```python
input_1 = np.random.randint(0, 10, size=(5, 5))

kernal_1 = np.array([
    [1, 0, -1],
    [1, 0, -1],
    [1, 0, -1]
])
```

```python
in_h, in_w = input_1.shape
k_h, k_w = kernal_1.shape
```

```python
out_h = in_h - k_h + 1
out_w = in_w - k_w + 1
```

```python
activation_map = np.zeros((out_h, out_w))
```

```python
for i in range(out_h):
    for j in range(out_w):
        region = input_1[i:i+k_h, j:j+k_w]
        activation_map[i, j] = np.sum(region * kernal_1)
```

```python
print("* 5x5 Input Matrix:")
print(input_1)
print("\n* 3x3 Kernel:")
print(kernal_1)
print("\n* Resulting 3x3 Activation Map:")
print(activation_map)
```

* 2. Implement a Python function apply_stride_convolution(input_matrix, kernel, stride) that performs convolution with a given stride on a 6x6 matrix and a 3x3 kernel, returning the resulting activation map.<br><br><em><strong>Hint:</strong> Use nested loops to move the kernel by the stride value over the input matrix.</em>

```python
def apply_stride_convolution(input_matrix, kernel, stride=1):
    in_h, in_w = input_matrix.shape
    k_h, k_w = kernel.shape
    
    out_h = (in_h - k_h) // stride + 1
    out_w = (in_w - k_w) // stride + 1
    
    activation_map = np.zeros((out_h, out_w))
    
    for i in range(out_h):
        for j in range(out_w):
            r_start = i * stride
            c_start = j * stride
            
            region = input_matrix[r_start:r_start+k_h, c_start:c_start+k_w]
            activation_map[i, j] = np.sum(region * kernel)
            
    return activation_map
```

```python
np.random.seed(10)
input_2 = np.random.randint(0, 10, size=(6, 6))
kernel_2 = np.array([
    [1, 1, 1],
    [0, 0, 0],
    [-1, -1, -1]
])
```

```python
output_stride_2 = apply_stride_convolution(input_2, kernel_2, stride=2)
```

```python
print("* 6x6 Input Matrix:")
print(input_2)
print("\n* Activation Map (Stride = 2):")
print(output_stride_2)
```

* 3. Modify your convolution code to add zero-padding of 1 around the input matrix before applying the convolution, and show the new activation map output.

Adding $P = 1$ zero-padding expands a $5 \times 5$ matrix to $7 \times 7$. Applying a $3 \times 3$ kernel with stride 1 returns an output equal in size to the original input ($5 \times 5$).

```python
def apply_padded_convolution(input_matrix, kernel, padding=1, stride=1):
    padded_input = np.pad(input_matrix, pad_width=padding, mode='constant', constant_values=0)
    return apply_stride_convolution(padded_input, kernel, stride=stride), padded_input
```

```python
input_3 = np.array([
    [6, 3, 7, 4, 6],
    [9, 2, 6, 7, 4],
    [3, 7, 7, 2, 5],
    [4, 1, 7, 5, 1],
    [4, 0, 9, 5, 8]
])
```

```python
kernel_3 = np.array([
    [0, 1, 0],
    [1, -4, 1],
    [0, 1, 0]
])
```

```python
padded_map, padded_matrix = apply_padded_convolution(input_3, kernel_3, padding=1, stride=1)
```

```python
print("* Padded Input Matrix (7x7):")
print(padded_matrix)
print("\n* New Activation Map Output (5x5):")
print(padded_map)
```
