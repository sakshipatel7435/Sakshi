* 1. Create a Python script using NumPy to perform a manual forward pass for a single-layer neural network: given an input vector X and a weight matrix W, compute the dot product and apply the sigmoid activation function to get the output.

```python
import numpy as np
```

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

X = np.array([0.5, -0.2, 0.8])

W = np.array([
    [0.1, 0.4],
    [-0.3, 0.7],
    [0.5, -0.6]
])

z = np.dot(X, W)
output = sigmoid(z)
```

```python
print("Input X:            ", X)
print("Linear Output (z):  ", np.round(z, 4))
print("Sigmoid Output (a): ", np.round(output, 4))
```

* 2. Modify your script to handle a two-layer neural network: take an input vector, compute the first hidden layer's activations using a weight matrix and ReLU activation, then compute the output layer's activations using a second weight matrix and sigmoid activation.

```python
def relu(x):
    return np.maximum(0, x)

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

X = np.array([1.2, -0.5, 0.3, -0.8])

W1 = np.array([
    [0.15, -0.2, 0.4],
    [0.8, 0.1, -0.3],
    [-0.5, 0.6, 0.2],
    [0.3, -0.4, 0.9]
])

W2 = np.array([
    [0.7],
    [-0.5],
    [0.3]
])

z1 = np.dot(X, W1)
a1 = relu(z1)

z2 = np.dot(a1, W2)
a2 = sigmoid(z2)
```

```python
print("Hidden Layer Activations (ReLU):   ", np.round(a1, 4))
print("Output Layer Activations (Sigmoid):", np.round(a2, 4))
```

* 3. Visualize the activations of each layer in your two-layer network using matplotlib — plot the input, hidden, and output activations as bar charts to show how data transforms through the network.<br><br><em><strong>Hint:</strong> Use plt.bar() for each layer and display them in separate subplots.</em>

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def relu(x):
    return np.maximum(0, x)

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

X = np.array([1.2, -0.5, 0.3, -0.8])
W1 = np.array([
    [0.15, -0.2, 0.4],
    [0.8, 0.1, -0.3],
    [-0.5, 0.6, 0.2],
    [0.3, -0.4, 0.9]
])
W2 = np.array([[0.7], [-0.5], [0.3]])

# Forward pass computations
a1 = relu(np.dot(X, W1))
a2 = sigmoid(np.dot(a1, W2))

fig, axes = plt.subplots(1, 3, figsize=(12, 4))
```

```python
axes[0].bar([f"X_{i}" for i in range(len(X))], X, color='skyblue')
axes[0].set_title('Input Vector (X)')
axes[0].set_ylabel('Values')
axes[0].axhline(0, color='black', linewidth=0.8, linestyle='--')

axes[1].bar([f"H_{i}" for i in range(len(a1))], a1, color='orange')
axes[1].set_title('Hidden Layer (ReLU)')
axes[1].set_ylabel('Activation')
axes[1].set_ylim(bottom=0)

axes[2].bar([f"Out_{i}" for i in range(len(a2))], a2, color='green')
axes[2].set_title('Output Layer (Sigmoid)')
axes[2].set_ylabel('Probability')
axes[2].set_ylim(0, 1)

plt.tight_layout()
plt.show()
```

* 4. Given a set of random weights and a batch of three input vectors, perform a forward pass through a three-layer network (input, one hidden, output) and print the output activations for each input.<br><br><em><strong>Constraint:</strong> Use only NumPy for all matrix operations and activations.</em>

```python
np.random.seed(42)

def relu(x):
    return np.maximum(0, x)

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

X_batch = np.array([
    [ 0.5, -0.3,  0.8,  0.1],
    [-0.2,  0.9, -0.5,  0.4],
    [ 1.0,  0.0, -0.7, -0.3]
])

W1 = np.random.randn(4, 5)

W2 = np.random.randn(5, 3)

Z1 = np.dot(X_batch, W1)
A1 = relu(Z1)

Z2 = np.dot(A1, W2)
A2 = sigmoid(Z2)

print("* Output Activations for Batch of 3:")
for i, sample_out in enumerate(A2):
    print(f"Sample {i+1}: {np.round(sample_out, 4)}")
```
