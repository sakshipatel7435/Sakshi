* 1. Write a Python function calculate_mse(predictions, targets) that takes two lists of numbers and returns the Mean Squared Error between them using the MSE formula.

$$\text{MSE} = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2$$

```python
import numpy as np
```

```python
def calculate_mse(predictions, targets):
    p = np.array(predictions)
    t = np.array(targets)
    return np.mean((p - t) ** 2)

preds = [2.5, 0.0, 2.1, 7.8]
targs = [3.0, -0.5, 2.0, 8.0]

print("MSE:", round(calculate_mse(preds, targs), 4))
```

* 2. Given two arrays: actual_ratings = [4, 3, 5, 2, 1] and predicted_ratings = [5, 2, 4, 2, 1], calculate the Mean Absolute Error (MAE) using Python and print the result.

$$\text{MAE} = \frac{1}{n} \sum_{i=1}^{n} \vert{}y_i - \hat{y}_i\vert{}$$

```python
actual_ratings = [4, 3, 5, 2, 1]
predicted_ratings = [5, 2, 4, 2, 1]

mae = np.mean(np.abs(np.array(actual_ratings) - np.array(predicted_ratings)))
print("Mean Absolute Error (MAE):", mae)
```

* 3. Implement a function binary_cross_entropy(y_true, y_pred) in Python that computes the binary cross-entropy loss for a batch of predictions and true labels (both as lists of 0s and 1s), then test it with y_true = [1, 0, 1] and y_pred = [0.9, 0.2, 0.7].

$$\text{BCE} = -\frac{1}{n} \sum_{i=1}^{n} \left[ y_i \log(\hat{y}_i) + (1 - y_i) \log(1 - \hat{y}_i) \right]$$

```python
def binary_cross_entropy(y_true, y_pred, eps=1e-15):
    y_true = np.array(y_true, dtype=float)
    y_pred = np.clip(np.array(y_pred, dtype=float), eps, 1 - eps)
    
    loss = -np.mean(y_true * np.log(y_pred) + (1 - y_true) * np.log(1 - y_pred))
    return loss

y_true = [1, 0, 1]
y_pred = [0.9, 0.2, 0.7]
```

```python
bce_loss = binary_cross_entropy(y_true, y_pred)
print("Binary Cross-Entropy Loss:", round(bce_loss, 4))
```

* 4. Simulate a basic Categorical Cross-Entropy calculation: given y_true = [0, 0, 1] and y_pred = [0.2, 0.3, 0.5], write Python code to compute the categorical cross-entropy loss for this single data point.

$$\text{CCE} = -\sum_{i=1}^{C} y_i \log(\hat{y}_i)$$

```python
y_true = np.array([0, 0, 1])
y_pred = np.array([0.2, 0.3, 0.5])

y_pred = np.clip(y_pred, 1e-15, 1 - 1e-15)   # Clip for safety against log(0)

cce_loss = -np.sum(y_true * np.log(y_pred))

print("Categorical Cross-Entropy Loss:", round(cce_loss, 4))
```

#### App: Zomato
Scenario: Predicting Estimated Time of Arrival (ETA) for Delivery
When predicting delivery time in minutes, choosing between MSE and MAE directly changes driver dispatch behavior:

Choice: Mean Absolute Error (MAE)

Why: Traffic logs and kitchen prep times frequently contain severe outliers (e.g., a massive rainstorm or a kitchen backup causing a 90-minute delay on a typical 20-minute delivery).

MSE squarings these errors, heavily penalizing extreme outliers. This pulls the model to artificially overestimate delivery times across all normal orders to compensate.

MAE treats errors linearly, making the model robust against extreme delivery anomalies so it continues providing accurate, realistic ETAs for everyday orders.
