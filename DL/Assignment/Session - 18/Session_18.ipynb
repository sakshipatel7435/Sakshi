* 2. Implement a simple 2D convolution operation in Python using NumPy: given a 5x5 grayscale image array and a 3x3 filter, compute the output feature map without using any deep learning libraries.<br><br><em><strong>Hint:</strong> Use nested loops to slide the filter over the image and sum the element-wise products.</em>

````python
image = np.array([
    [1, 2, 3, 0, 1],
    [0, 1, 2, 4, 1],
    [3, 1, 0, 2, 0],
    [1, 0, 1, 3, 2],
    [2, 1, 0, 1, 1]
])
````

````python
filter_kernel = np.array([
    [ 1,  0, -1],
    [ 1,  0, -1],
    [ 1,  0, -1]
])
````

````python
img_h, img_w = image.shape
f_h, f_w = filter_kernel.shape
````

````python
out_h = img_h - f_h + 1
out_w = img_w - f_w + 1
````

````python
output_feature_map = np.zeros((out_h, out_w))
````

````python
for i in range(out_h):
    for j in range(out_w):
        region = image[i:i+f_h, j:j+f_w]

        output_feature_map[i, j] = np.sum(region * filter_kernel)
````

````python
print("* 5x5 Input Image:")
print(image)
print("\n* 3x3 Kernel:")
print(filter_kernel)
print("\n* Output Feature Map (3x3):")
print(output_feature_map)
````

**Component Descriptions** 
* Input Layer: Receives the raw numerical pixel values of the image (e.g., $28 \times 28 \times 3$ array for RGB) and passes them into the network.

* Convolutional Layer: Applies sliding filters across the input to perform element-wise multiplication, extracting spatial features like edges, textures, and lines into feature maps.
* Pooling Layer: Downsamples feature maps (e.g., using $2 \times 2$ max pooling) to reduce spatial dimensions, lower computational costs, and prevent overfitting.
* Output Layer: Flattens spatial feature vectors and applies fully connected neurons with an activation function (like Softmax) to compute final class probability scores.

* 4. Write a function in Python that takes an image array and returns a new array where each pixel is replaced by the maximum value in its 2x2 local neighborhood (simulate max pooling).<br><br><em><strong>Constraint:</strong> Do not use any deep learning libraries like TensorFlow or PyTorch for this task.</em>

````python
def max_pooling_2d(image_array, pool_size=(2, 2), stride=2):
    img_h, img_w = image_array.shape
    p_h, p_w = pool_size
    
    out_h = (img_h - p_h) // stride + 1
    out_w = (img_w - p_w) // stride + 1
    
    pooled_output = np.zeros((out_h, out_w), dtype=image_array.dtype)
    
    for i in range(out_h):
        for j in range(out_w):
            row_start = i * stride
            col_start = j * stride
            
            region = image_array[row_start:row_start+p_h, col_start:col_start+p_w]
            
            pooled_output[i, j] = np.max(region)
            
    return pooled_output
````

````python
sample_image = np.array([
    [1, 3, 2, 4],
    [5, 6, 1, 2],
    [8, 7, 0, 1],
    [2, 4, 3, 5]
])
````

````python
result = max_pooling_2d(sample_image)
````

````python
print("* Original 4x4 Array:")
print(sample_image)
print("\n* Max Pooled 2x2 Array:")
print(result)
````

* 5. Explain with a real-world example from Instagram or Zomato how translation invariance in CNNs is useful. Write 2-3 sentences connecting it to how these apps recognize faces or food items even if they appear in different positions in the image.

**Real-World Example — Translation Invariance**

When users upload food photos to Zomato or selfies to Instagram, a pizza or face might appear centered, off-to-the-side, or in a corner of the image frame.

**Translation invariance** ensures that because CNNs use sliding shared-weight filters and spatial reduction (pooling layers), they identify key localized patterns—such as the circular crust of a pizza or facial features—regardless of where those features are located in the pixel matrix. Consequently, these applications can correctly identify and auto-tag objects or faces even if their position shifts across different images.
