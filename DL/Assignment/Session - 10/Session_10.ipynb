### Tasks

* 1. Create a simple Keras Sequential model for classifying handwritten digits (use the MNIST dataset), and train it without any regularization. Save the validation accuracy after 5 epochs.

```python
import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense, Flatten
```

```python
mnist = tf.keras.datasets.mnist
```

```python
(X_train, y_train), (X_val, y_val) = mnist.load_data()
```

```python
X_train = X_train/255.0
X_val = X_val/255.0
```

```python
model_1 = Sequential([
    Flatten(input_shape = (28,28)),
    Dense(128, activation='relu'),
    Dense(64, activation='relu'),
    Dense(10, activation='softmax')
])
```

```python
model_1.compile(
    optimizer = 'adam',
    loss = 'sparse_categorical_crossentropy',
    metrics = ['accuracy']
)
```

```python
history_1 = model_1.fit(
    X_train, y_train, 
    validation_data=(X_val, y_val), 
    epochs=5, 
    batch_size=64)
```

```python
val_accuracy_1 = history_1.history['val_accuracy'][-1]
print(f"Validation Accuracy: {val_accuracy_1:.4f}")
```

* 2. Modify your Keras model to add L2 regularization (kernel_regularizer=regularizers.l2(0.01)) to all Dense layers, retrain for 5 epochs, and compare the validation accuracy to your previous result.

```python
from tensorflow.keras import regularizers
```

```python
model_2 = Sequential([
    Flatten(input_shape = (28,28)),
    Dense(128, activation='relu', kernel_regularizer=regularizers.l2(0.01)),
    Dense(64, activation='relu', kernel_regularizer=regularizers.l2(0.01)),
    Dense(10, activation='softmax')
])
```

```python
model_2.compile(
    optimizer = 'adam',
    loss = 'sparse_categorical_crossentropy',
    metrics = ['accuracy']
)
```

```python
history_2 = model_2.fit(
    X_train, y_train, 
    validation_data=(X_val, y_val), 
    epochs=5, 
    batch_size=64)
```

```python
val_accuracy_2 = history_2.history['val_accuracy'][-1]
print(f"Validation Accuracy: {val_accuracy_1:.4f}")
print(f"Validation Accuracy (L2 Regulization): {val_accuracy_1:.4f}")
```

* 3. Add Dropout layers (with rate=0.5) after each Dense layer in your Keras model, retrain for 5 epochs, and observe the change in validation accuracy.<br><br><em><strong>Hint:</strong> Use keras.layers.Dropout and place it after activations.</em>

```python
from tensorflow.keras.layers import Dropout
```

```python
model_3 = Sequential([
    Flatten(input_shape = (28,28)),
    Dense(128, activation='relu'),
    Dropout(0.5),
    Dense(64, activation='relu'),
    Dropout(0.5),
    Dense(10, activation='softmax')
])
```

```python
history_3 = model_3.fit(
    X_train, y_train, 
    validation_data=(X_val, y_val), 
    epochs=5, 
    batch_size=64)
```

```python
val_accuracy_3 = history_3.history['val_accuracy'][-1]
print(f"Validation Accuracy (Dropout): {val_accuracy_3:.4f}")
```

* 5. Implement EarlyStopping in your model training using keras.callbacks.EarlyStopping with patience=2, and show how it affects the number of training epochs and final validation accuracy.

```python
from tensorflow.keras.callbacks import EarlyStopping
```

```python
model_5 = Sequential([
    Flatten(input_shape = (28,28)),
    Dense(128, activation='relu'),
    Dense(64, activation='relu'),
    Dense(10, activation='softmax')
])
```

```python
model_5.compile(
    optimizer = 'adam',
    loss = 'sparse_categorical_crossentropy',
    metrics = ['accuracy']
)
```

```python
early_stop = EarlyStopping(
    monitor = 'val_loss',
    patience = 2,
    restore_best_weights = True
)
```

```python
history_5 = model_5.fit(
    X_train, y_train, 
    validation_data=(X_val, y_val), 
    epochs=20, 
    batch_size=64,
    callbacks=[early_stop]
)
```

```python
actual_epochs = len(history_5.history['loss'])
final_val_acc = history_5.history['val_accuracy'][-1]

print(f"\nTraining stopped automatically at epoch: {actual_epochs}")
print(f"Final Validation Accuracy: {final_val_acc:.4f}")
```
