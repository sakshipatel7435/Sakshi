* Low Learning Rate (e.g., 0.001): The decision boundary changes very slowly and smoothly. It takes a high number of epochs to find the right shape to separate the orange and blue dots. While it is slow, it is highly stable and steadily converges on a good solution.

Takeaway: A moderate learning rate (like 0.03 or 0.1) usually converges the fastest and most smoothly for simple classification tasks without overshooting or stalling.

* 4. Write a step-by-step workflow for building a deep learning model for movie recommendation, starting from collecting user data to deploying the trained model in a real app.<br><br><em><strong>Hint:</strong> Use the stages: Data → Model → Training → Evaluation → Deployment.</em>

### Deep Learning Workflow: Movie Recommendation System

1.  **Data Collection:** Gather user watch history, ratings, and clicks. Clean the data and split it into training and testing sets.

2.  **Data Preprocessing:** Handle missing ratings, encode movie genres into numerical arrays, scale watch times, and split the data into training, 
validation, and testing sets.

3.  **Model:** Create a neural network that converts users and movies into "embeddings" (numerical profiles) to map out user preferences and movie genres.

4.  **Training:** Feed the data to the model using an optimizer (like Adam) to minimize prediction errors and teach it what movies users like.

5.  **Evaluation:** Test the model using metrics like RMSE or Precision@K to ensure the top-recommended movies are actually accurate.

6.  **Deployment:** Host the trained model on a server API. The mobile app calls this API to display personalized movie lists to the user instantly.

7.  **Monitoring:** Continuously track the model's performance in production. If the click-through rate on recommended movies drops or user preferences shift (data/concept drift), automatically trigger a pipeline to retrain the model with fresh data.
