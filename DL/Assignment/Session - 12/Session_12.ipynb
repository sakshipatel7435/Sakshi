* 2. Create a Sequential model in Keras with three layers: an input Dense layer with 32 units and 'relu' activation, a Dense layer with 16 units and 'relu', and an output Dense layer with 1 unit and 'sigmoid'. Print the model summary.

```python
import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense, Input
from tensorflow.keras.models import Model
```

```python
model_1 = Sequential([
    Input(shape=(10,)),
    Dense(32, activation='relu', name='dense_1'),
    Dense(16, activation='relu', name='dense_2'),
    Dense(1, activation='sigmoid', name='output_layer')
])

model_1.summary()
```

* 3. Build a simple Keras Functional API model that takes a 10-feature input, passes it through one Dense layer with 20 units and 'relu', then outputs a single value with a Dense layer. Display the model summary and total parameter count.

```python
inputs = Input(shape=(10,), name='input_layer')
x = Dense(20, activation='relu', name='hidden_layer')(inputs)
outputs = Dense(1, name='output_layer')(x)

model_3 = Model(inputs=inputs, outputs=outputs, name='functional_model')

model_3.summary()

print(f"Total Parameter Count: {model_3.count_params()}")
```

* 5. Use ChatGPT or Copilot to generate Keras code for a model with two inputs (e.g., one for user age and one for number of songs listened on Spotify) and one output (predicting likelihood of buying a premium subscription). Paste the generated code, run it, and note any errors or changes you had to make.

```python
import numpy as np
from tensorflow.keras.layers import concatenate
```

```python
age_input = Input(shape=(1,), name='age_input')
songs_input = Input(shape=(1,), name='songs_listened_input')

combined = concatenate([age_input, songs_input])
```

```python
x = Dense(16, activation='relu')(combined)
x = Dense(8, activation='relu')(x)
output = Dense(1, activation='sigmoid', name='subscription_prob')(x)
```

```python
spotify_model = Model(inputs=[age_input, songs_input], outputs=output)
```

```python
spotify_model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
spotify_model.summary()
```

```python
dummy_age = np.array(([24], [45], [19]))
dummy_songs = np.array(([150], [10], [500]))
dummy_target = np.array(([1], [0], [1]))

spotify_model.train_on_batch([dummy_age, dummy_songs], dummy_target)
```
