* 1. Train a simple image classifier on the Fashion MNIST dataset using Keras and the SGD optimizer with default parameters, then record the accuracy and loss after 5 epochs.

```python
import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense, Flatten
from tensorflow.keras.optimizers import SGD
```

```python
fashion_mnist = tf.keras.datasets.fashion_mnist
```

```python
(X_train, y_train), (X_val, y_val) = fashion_mnist.load_data()
```

```python
X_train = X_train/255.0
X_val = X_val/255.0
```

```python
def create_model():
    return Sequential([
        Flatten(input_shape = (28, 28)),
        Dense(128, activation='relu'),
        Dense(64, activation='relu'),
        Dense(10, activation='softmax')
    ])
```

```python
model_1 = create_model()

model_1.compile(
    optimizer = SGD(),
    loss = 'sparse_categorical_crossentropy',
    metrics = ['accuracy']
)
```

```python
history_1 = model_1.fit(
    X_train, y_train,
    validation_data = (X_val, y_val),
    epochs = 5,
    batch_size = 64
)
```

```python
val_loss_1 = history_1.history['val_loss'][-1]
val_accuracy_1 = history_1.history['val_accuracy'][-1]

print(f"Validation Loss (SGD): {val_loss_1:.4f}\nValidation Accuracy (SGD): {val_accuracy_1:.4f}")
```

* 4. Implement a learning rate schedule in your Keras model using the LearningRateScheduler callback to reduce the learning rate by half after every 2 epochs, and train with the Momentum optimizer; report how the learning rate changes during training.

```python
from tensorflow.keras.callbacks import LearningRateScheduler
from tensorflow.keras.optimizers import SGD
```

```python
def lr_decay(epoch, lr):
    if epoch > 0 and epoch % 2 == 0:
        return lr * 0.5
    return lr
```

```python
model_4 = create_model()

model_4.compile(
    optimizer = SGD(learning_rate=0.01, momentum=0.9),
    loss = 'sparse_categorical_crossentropy',
    metrics = ['accuracy']
)
```

```python
lr_callback = LearningRateScheduler(lr_decay, verbose = 1)
```

```python
history_4 = model_4.fit(
    X_train, y_train,
    validation_data = (X_val, y_val),
    epochs = 5,
    batch_size = 64,
    callbacks = [lr_callback]
)
```

**Overview:**

The learning rate schedule was implemented using LearningRateScheduler with the decay condition lr * 0.5 triggered every 2 epochs. The model was trained using the SGD optimizer with Momentum (0.9) starting from an initial learning rate of 0.01.

**Learning Rate Changes During Training**
* Epoch 1: lr = 0.0100 (0.009999999776...)
Status: Training begins with the initial base learning rate.

* Epoch 2: lr = 0.0100 (0.009999999776...)
Status: Retained at 0.01. (Decay condition epoch % 2 == 0 is false for 0-indexed epoch 1).

* Epoch 3: lr = 0.0050 (0.004999999888...)
Status: Halved by 50% from 0.01 to 0.005.

* Epoch 4: lr = 0.0050 (0.004999999888...)
Status: Retained at 0.005.

* Epoch 5: lr = 0.0025 (0.002499999944...)
Status: Halved by 50% again from 0.005 to 0.0025.

**Summary & Observation**

Progression: The learning rate systematically scaled down as expected: 0.0100 ➔ 0.0100 ➔ 0.0050 ➔ 0.0050 ➔ 0.0025.

**Model Impact:** As the learning rate decreased, the validation loss dropped continuously from 0.4704 (Epoch 1) down to 0.3645 (Epoch 5), while validation accuracy steadily improved from 83.27% to 86.72%. Step-down decay helped the Momentum optimizer stabilize and converge smoothly into lower loss minima without overshooting.

* 5. Use ChatGPT to explain, in your own words, why Adam is often preferred in industry projects over SGD or RMSProp, and give one real-world app scenario (like Zomato, Flipkart, or Spotify) where optimizer choice could impact user experience.

#### Why Adam is Preferred:
* Adam combines the best parts of Momentum (accelerating past noisy gradients) and RMSprop (adapting individual learning rates for each parameter based on recent gradient history). In real-world projects, data can be sparse, noisy, or non-stationary. Adam handles these conditions reliably out of the box without requiring manual tuning of learning rate schedules across thousands of parameters.

#### Real-World Impact (Spotify Recommender System):
* Scenario: Spotify trains deep collaborative filtering models to recommend songs in real time based on user listens, skips, and search terms.

#### Impact of Optimizer:
* If standard SGD is used, updating sparse features (e.g., niche genres or rare artists) takes millions of iterations because non-adaptive learning rates treat every feature equally, leading to slow updates for rarely streamed songs.

* Adam scales updates inversely to gradient frequency, allowing the model to quickly learn strong embeddings for niche songs. This directly prevents poor song recommendations on a user's personalized homepage.
