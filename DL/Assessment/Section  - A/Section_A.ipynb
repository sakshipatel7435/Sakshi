### 3. SCENARIO
You are deploying a food-delivery-time prediction model for a logistics company. The
trained scikit-learn model is saved as a .pkl file, and the Streamlit app must let
dispatchers enter order details and receive a predicted delivery time instantly without
the model reloading on every interaction.

* Question: Describe the sequence of steps — from saving the model file to displaying the
prediction — that makes this Streamlit prediction app work correctly, and explain why wrapping
the model load function with st.cache_resource is critical for production performance.

### Ans:
* **Sequence**: Model saved with joblib.dump(model, 'file.pkl') $\rightarrow$ Streamlit app loads model using joblib.load() $\rightarrow$ App captures user inputs from UI $\rightarrow$ Transforms inputs into NumPy array $\rightarrow$ Calls model.predict() $\rightarrow$ Displays result using st.success(). 

* **Why st.cache_resource**: Streamlit re-executes the entire script from top to bottom on every user interaction. Without caching, the heavy model loads from disk repeatedly, causing lag or memory leaks.

### 4. SCENARIO
You are building a neural network to classify incoming food orders into three urgency
tiers — Standard, Priority, and Express — based on order value, distance from restaurant,
and estimated preparation time. Your current model uses a sigmoid activation on the
output layer.


* Question: Which activation function should the output layer use for this three-class
classification task, and why does using sigmoid on the output layer produce incorrect probability
distributions when you need mutually exclusive class probabilities across all three tiers?

### Ans:
* Activation: Softmax activation.  
* Why Sigmoid Fails: Sigmoid treats each output node independently (probabilities sum to $>1$ or $<1$). Softmax normalizes exponents so all output probabilities sum to exactly $1.0$, creating mutually exclusive class probabilities needed for 3 urgency tiers. 

### 5. SCENARIO
You are training an ANN on a food delivery platform's dataset to predict peak-hour
demand. After 30 epochs the training accuracy reaches 96% but validation accuracy
stalls at 71%, indicating severe overfitting. Your model has three Dense hidden layers.

* Question: Compare Dropout regularization and L2 weight regularization as strategies to reduce
this gap. For a dense multi-layer neural network, which would you apply first in practice, at what
typical rate or strength, and what different mechanism does each use to discourage overfitting?


### Ans:
* **Mechanism**: Dropout randomly zeroes out nodes per step to prevent co-adaptation. L2 adds a penalty proportional to squared weight magnitudes ($\vert{}w\vert{}^2$) to loss, keeping weights small.  
* **Practical Order**: Apply Dropout first (rate $0.2 - 0.5$) as it is generally more effective for deep fully connected layers. 

### 6. SCENARIO
You are designing a system for a food delivery platform to automatically classify
customer-submitted food photos as 'acceptable quality' or 'poor quality'. A colleague
suggests using a simple ANN with pixel values flattened into a 1D input vector. You are
considering a CNN instead.

* Question: Explain the key architectural difference between an ANN and a CNN that makes CNNs
significantly more effective for image classification, and describe what a convolutional filter
learns from spatial data that fully connected layers cannot efficiently capture

### Ans:
* **Key Difference**: Flattening images in ANNs destroys spatial relationships (3D structure). CNNs maintain 2D/3D spatial grids. 
* **Convolutional Filters**: Convolutional filters learn localized spatial patterns (edges, textures, shapes) using weight sharing, making them translation-invariant and drastically reducing parameter counts compared to fully connected layers.
