* 2. Explain in your own words how an LLM like ChatGPT generates a reply when you ask it a question on WhatsApp or a website chatbot. Write your answer in 4-5 sentences, focusing on how it understands your input and produces a response.

----

#### Ans:

When you send a prompt to an LLM via WhatsApp or a web interface, the model first breaks down your input into smaller units called tokens (words or sub-words) and converts them into numerical vectors (embeddings). Next, its Transformer neural network processes these vectors through its attention mechanisms to map the context and relationships between your words. Instead of searching a database for pre-written answers, the LLM uses its statistical knowledge gained from pre-training to predict the most logically coherent next token sequentially. Finally, these continuous token predictions are decoded back into readable text and streamed back to your screen in real time.

### Ans:


**My Prompt:**
> What are two advantages and two limitations of using LLMs in business applications?

**ChatGPT's Answer:**
> **Advantages:**
> 
> 1. *Increased Efficiency and Automation:* LLMs can automate repetitive text-based tasks like drafting emails, summarizing reports, and providing 24/7 customer support.
> 2. *Scalability:* A single model can handle thousands of customer queries simultaneously, allowing businesses to scale operations without massively increasing headcount.
> 
> **Limitations:**
> 
> 1. *Hallucinations and Accuracy:* LLMs can confidently invent incorrect facts, which is dangerous in fields like legal or medical business where accuracy is critical.
> 2. *Data Privacy and Security Risks:* Inputting sensitive corporate data or customer PII into a public LLM can lead to severe data leaks if the model is not properly secured.

**My Opinion:** I strongly agree with the limitation regarding **Hallucinations**. In a business environment, trust and accuracy are everything. If an e-commerce chatbot confidently invents a fake return policy or hallucinates a non-existent 50% discount code for a customer, the business is forced to honor it or risk a massive PR disaster. This proves that companies cannot just plug an LLM into their systems blindly without strict guardrails in place.
