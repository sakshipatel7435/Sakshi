### Tasks

* 1. Install Streamlit using pip, create a new file called insta_caption_generator.py, and display the text 'Welcome to Insta Caption Generator!' on the main page.

* 2. Build a simple Streamlit UI in insta_caption_generator.py with a text input labeled 'Describe your photo' and a button labeled 'Generate Caption'.

* 3. When the 'Generate Caption' button is clicked, display the user's input below the button with the label 'Your Description:'.

* 4. Connect your Streamlit app to the OpenAI GPT API (or Hugging Face Inference API) so that when a user enters a description and clicks 'Generate Caption', the app fetches an AI-generated Instagram caption and displays it below.<br><br><em><strong>Hint:</strong> Use the requests library to call the API and handle the response.</em>

* 5. Extend your app to show a simple chat interface: let the user enter multiple descriptions one after another, and display the last 3 AI-generated captions as chat bubbles (user input and AI response) on the page.
