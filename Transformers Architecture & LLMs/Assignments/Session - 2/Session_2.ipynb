* 1. Install the transformers library in your Python environment and write a script that loads the GPT-2 model and generates a short Instagram-style caption based on the prompt 'Sunset vibes at the beach'.

```python
# Installation: Run this in your terminal or notebook environment
# !pip install transformers torch
```

```python
from transformers import pipeline
import warnings
warnings.filterwarnings("ignore")

caption_generator = pipeline('text-generation', model='gpt2')

prompt = "Sunset vibes at the beach"
output = caption_generator(
    prompt, 
    max_new_tokens=15,
    num_return_sequences=1, 
    pad_token_id=50256
)

print("\n\nGenerated Caption:\n", output[0]['generated_text'])
```

Use the transformers library to load the BERT-base-uncased model and write code to classify the sentiment (positive or negative) of the following Zomato review: 'The food was absolutely delicious and the delivery was super fast.'<br><br><em><strong>Hint:</strong> Use the pipeline('sentiment-analysis') function for quick setup.</em>

```python
from transformers import pipeline

sentiment_analyzer = pipeline('sentiment-analysis', model='distilbert-base-uncased-finetuned-sst-2-english')

review = "The food was absolutely delicious and the delivery was super fast."
result = sentiment_analyzer(review)

print("Review Sentiment Analysis Result:")
print(f"Label: {result[0]['label']} | Score: {result[0]['score']:.4f}")
```

* 3. Load the T5-base model using the transformers library and write a script that paraphrases the sentence 'BookMyShow makes movie ticket booking easy and fast.'<br><br><em><strong>Hint:</strong> Use the prefix 'paraphrase: ' before your input text for T5.</em>

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

tokenizer = AutoTokenizer.from_pretrained("t5-base", legacy=False)
model = AutoModelForSeq2SeqLM.from_pretrained("t5-base")

text = "BookMyShow makes movie ticket booking easy and fast."
prompt = f"paraphrase: {text}"

input_ids = tokenizer(prompt, return_tensors="pt").input_ids
outputs = model.generate(input_ids, max_new_tokens=50)
generated_text = tokenizer.decode(outputs[0], skip_special_tokens=True)

print(text)
print(f"T5 Transformed Output: {generated_text}")
```

* 5. Given the following scenarios, decide which model (GPT, BERT, or T5) is best suited and explain your reasoning in 1-2 lines for each: 
    * 1) Generating a playlist description for Spotify, 
    * 2) Classifying spam messages on WhatsApp, 
    * 3) Summarizing a cricket match report for an IPL app.

----

### Ans:

1. **Generating a playlist description for Spotify**

   * **Best Model:** GPT
   * **Reasoning:** GPT is a Decoder-only generative model, making it perfectly suited for creative, open-ended text generation tasks like writing engaging descriptions from scratch.

2. **Classifying spam messages on WhatsApp**

   * **Best Model:** BERT
   * **Reasoning:** BERT is an Encoder-only model designed to deeply understand the bidirectional context of a sentence, making it highly accurate for analytical text classification tasks like spam detection.

3. **Summarizing a cricket match report for an IPL app**

   * **Best Model:** T5
   * **Reasoning:** T5 uses a robust Encoder-Decoder architecture which is explicitly designed for sequence-to-sequence transformation tasks, making it ideal for taking a long article (input sequence) and transforming it into a concise summary (output sequence).
