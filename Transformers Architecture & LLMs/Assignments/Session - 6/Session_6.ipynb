* 2. Create a Customer Support Chatbot flow for a food delivery app (like Zomato or Swiggy) using Python — the bot should answer three types of queries: order status, refund request, and delivery delay, using if-else logic based on user input.

```python
def food_delivery_chatbot(user_query):
    query = user_query.lower()
    
    if "status" in query or "where is" in query:
        return "Bot: Your order is currently out for delivery and will reach you in approximately 10 minutes!"
        
    elif "refund" in query or "money back" in query:
        return "Bot: I apologize for the inconvenience. A full refund has been initiated to your original payment method and will reflect within 3-5 business days."
        
    elif "delay" in query or "late" in query:
        return "Bot: We are experiencing heavy rain in your area which has caused a slight delay. The delivery partner is riding safely and will be there soon!"
        
    else:
        return "Bot: I'm sorry, I didn't quite catch that. Could you please specify if you need help with your 'order status', a 'refund', or a 'delivery delay'?"

print(food_delivery_chatbot("Where is my pizza? I need the order status."))
print(food_delivery_chatbot("My food was completely spilled, I want a refund."))
print(food_delivery_chatbot("Why is my delivery so late today?"))
```

* 3. Design an AI Content Generator function that takes a topic (e.g., 'Best Cafes in Ahmedabad') as input and uses OpenAI's GPT API to generate a short blog post about it.<br><br><em><strong>Hint:</strong> Use the openai.Completion.create() method and set max_tokens to 100 for a short output.</em>

```python
import requests

def generate_blog_post(topic):
    url = "https://api.groq.com/openai/v1/chat/completions"
    headers = {
        "Authorization": "Bearer API_KEY_HERE",
        "Content-Type": "application/json"
    }
    payload = {
        "model": "openai/gpt-oss-20b",
        "messages": [{"role": "user", "content": f"Write a short, engaging blog post about: {topic}"}],
        "max_tokens": 1000
    }
    
    try:
        response = requests.post(url, headers=headers, json=payload)
        if response.status_code != 200:
            return f"[SIMULATED OUTPUT due to missing API Key] \nWelcome to the ultimate guide on {topic}! Ahmedabad has an incredibly vibrant cafe culture perfect for college students and remote workers alike. Grab a cold coffee and enjoy the beautiful aesthetics!"
        
        result = response.json()
        return result['choices'][0]['message']['content'].strip()
    except Exception:
        return f"[SIMULATED OUTPUT due to missing API Key] \nWelcome to the ultimate guide on {topic}! Ahmedabad has an incredibly vibrant cafe culture perfect for college students and remote workers alike. Grab a cold coffee and enjoy the beautiful aesthetics!"

print("Generated Blog Post:")
print(generate_blog_post("Best Cafes in Ahmedabad"))
```

* 5. Use ChatGPT or Copilot to suggest improvements for your AI Content Generator function from Task 3, and implement one suggestion in your code.<br><br><em><strong>Hint:</strong> Ask the AI tool: 'How can I make my GPT-based content generator more engaging for users?'</em>

```python
def generate_engaging_blog_post(topic, tone="enthusiastic"):
    url = "https://api.groq.com/openai/v1/chat/completions"
    headers = {
        "Authorization": "Bearer API_KEY_HERE",
        "Content-Type": "application/json"
    }
    prompt = f"Act as an expert travel blogger. Write a short, highly {tone} and engaging blog post about: {topic}"
    payload = {
        "model": "openai/gpt-oss-20b",
        "messages": [{"role": "user", "content": prompt}],
        "max_tokens": 1000,
        "temperature": 0.8
    }
    
    try:
        response = requests.post(url, headers=headers, json=payload)
        if response.status_code != 200:
            return f"[SIMULATED {tone.upper()} OUTPUT] \nOMG guys, you absolutely HAVE to check out the {topic}! The vibe is literally unmatched and it's the perfect aesthetic for your Instagram feed!"
        
        result = response.json()
        return result['choices'][0]['message']['content'].strip()
    except Exception:
        return f"[SIMULATED {tone.upper()} OUTPUT] \nOMG guys, you absolutely HAVE to check out the {topic}! The vibe is literally unmatched and it's the perfect aesthetic for your Instagram feed!"

print("Enhanced Blog Post:")
print(generate_engaging_blog_post("Best Cafes in Ahmedabad", tone="Gen-Z enthusiastic"))
```
