4. For a Pandas Data Frame created from a NumPy array, what is the default behavior for the labels for the columns? For the rows? 


When you create a Pandas DataFrame from a NumPy array without specifying any extra arguments, Pandas follows a simple, zero-indexed logic for both dimensions.

The Default Behavior:
    For Columns: Pandas assigns a RangeIndex starting from 0 and increasing by 1 for each column.

    For Rows (Index): Pandas assigns a RangeIndex starting from 0 and increasing by 1 for each row.

```python
import pandas as pd
import numpy as np

arr = np.array([[10, 20, 30], 
                [40, 50, 60], 
                [70, 80, 90]])

df = pd.DataFrame(arr)
print(df)
```

10. Create an array of numbers between 0 and 6 with increments of 0.3 and name its “x”.
Then on the same plot, plot x, x², x³, and x⁴. For consistency, use the following style lines
respectively, “ro”, “bs”, “g”, and “:”. Lastly, make sure that the x-axis covers 0 to 6, while
the y-axis spans from 0 to 125. Do not worry if you are not familiar with the style lines —
you will recognize them as soon as you see the plot. 

```python
x = np.arange(0, 6, 0.3)


plt.plot(x, x**1, "ro")
plt.plot(x, x**2, "bs")
plt.plot(x, x**3, "g")
plt.plot(x, x**4, ":")

plt.xlim(0, 6)
plt.ylim(0, 125)

plt.xlabel("X")
plt.ylabel("Y")

plt.show()
```

12. Plot a histogram of x, where x consists of 100,000 randomly-selected points with a normal
distribution (hint: you can use numpy.random.randn() to generate the random points). The
histogram should have 10 bins. Look at how the histogram changes when we try 20 and 50
bins.

```python
import numpy as np
import matplotlib.pyplot as plt 

x = np.random.rand(100000)

plt.hist(x, bins=10)
plt.title("Histrogram with 10 bins")
plt.show()

plt.hist(x, bins=20)
plt.title("Histogram with 20 bins")
plt.show()

plt.hist(x, bins=50)
plt.title("Histogram with 30 bins")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.random.randn(100000)

plt.hist(x, bins=10)
plt.title("Histogram with 10 bins")
plt.show()

plt.hist(x, bins=20)
plt.title("Histogram with 20 bins")
plt.show()

plt.hist(x, bins=50)
plt.title("Histogram with 50 bins")
plt.show()
```
