### Task 3: Packaging Label Transformer and Edge Scanner

#### Build a program that normalises a food packaging label image through a chain of transformations and then highlights its edges for OCR readiness.

* **Requirement 1:** Load a food packaging image, rotate it 90 degrees clockwise using cv2.getRotationMatrix2D() and cv2.warpAffine(), then crop the lower-right quadrant of the rotated result using NumPy slicing.

```python
import cv2
```

```python
img = cv2.imread('food_packaging.jpg')
h, w = img.shape[:2]
```

```python
M = cv2.getRotationMatrix2D((w // 2, h // 2), -90, 1.0)
rotated = cv2.warpAffine(img, M, (w, h))
```

* **Requirement 2:** Flip the cropped region horizontally using cv2.flip() and display the result after each step — rotate, crop, and flip — in three separate named windows labelled clearly.

```python
rh, rw = rotated.shape[:2]
cropped = rotated[rh // 2: rh, rw // 2: rw]
```

```python
flipped = cv2.flip(cropped, 1)
```

```python
cv2.imshow("1 - Rotated", rotated)
cv2.imshow("2 - Cropped", cropped)
cv2.imshow("3 - Flipped", flipped)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

* **Requirement 3:** Apply a Gaussian blur (kernel size 5×5, sigmaX=0) to the final flipped image using cv2.GaussianBlur() to suppress noise before edge analysis.

```python
blurred = cv2.GaussianBlur(flipped, (5, 5), 0)
cv2.imshow("Gaussian Blur Image", blurred)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

* **Requirement 4:** Run Canny edge detection on the blurred image using cv2.Canny() with threshold values of your choice, display the edge map in a window titled 'Label Edges', and save it as label_edges.jpg.

```python
edges = cv2.Canny(blurred, 50, 150)
cv2.imshow("Label Edges", edges)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
