### Task 4: Live Delivery Agent Face Verifier

#### Build a real-time face detection program using a webcam that identifies a delivery agent at the door, draws a bounding box around the detected face, and captures a snapshot on keypress.

* **Requirement 1:**  Open the default webcam using cv2.VideoCapture(0) and read frames continuously in a loop, displaying each frame live in a window titled 'Delivery Agent Verifier'.

* **Requirement 2:**  Load the Haar cascade classifier for frontal face detection and apply detectMultiScale() on the grayscale version of each frame with minNeighbors=6 and minSize=(80, 80) to minimise false positives.

* **Requirement 3:**  Draw a green rectangle around every detected face using cv2.rectangle() and overlay the text 'Agent Detected' directly above the bounding box using cv2.putText() with a readable font size.

* **Requirement 4:**  When the user presses 's', save the current frame as agent_snapshot.jpg and print the save
confirmation to the console; when the user presses 'q', release the capture object and close all
windows cleanly.

```python
import cv2

cascade_path = cv2.data.haarcascades + 'haarcascade_frontalface_default.xml'
face_cascade = cv2.CascadeClassifier(cascade_path)

if face_cascade.empty():
    print("Error: Could not load Haar cascade classifier.")
else:
    cap = cv2.VideoCapture(0)
    
    if not cap.isOpened():
        print("Error: Could not open webcam.")
    else:
        while True:
            ret, frame = cap.read()
            if not ret:
                break
                
            gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
            
            faces = face_cascade.detectMultiScale(gray, scaleFactor=1.1, minNeighbors=6, minSize=(80, 80))
            
            for (x, y, w, h) in faces:
                cv2.rectangle(frame, (x, y), (x+w, y+h), (0, 255, 0), 2)
                cv2.putText(frame, 'Agent Detected', (x, y-10), cv2.FONT_HERSHEY_SIMPLEX, 0.9, (0, 255, 0), 2)
                
            cv2.imshow('Delivery Agent Verifier', frame)
            
            key = cv2.waitKey(1) & 0xFF
            if key == ord('s'):
                cv2.imwrite('agent_snapshot.jpg', frame)
                print("Snapshot saved as 'agent_snapshot.jpg'.")
            elif key == ord('q'):
                break
        
        cap.release()
        cv2.destroyAllWindows()
```
