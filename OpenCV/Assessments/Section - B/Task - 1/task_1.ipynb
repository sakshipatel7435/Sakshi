### Task 1: Food Image Loader and Inspector

##### Build a program that loads a food image from disk using OpenCV and prints its basic properties to the console.

* **Requirement 1:** Load any food image using cv2.imread() and verify it was read successfully — print a clear error message and exit gracefully if the file is not found.

```python
import cv2
import sys
```

```python
image = ""
img = cv2.imread(image)
```

```python
img = cv2.imread(r"food_image.jpg")
```

```python
if img is None:
    print(f"Error: Could not load image from '{img}. Check filename/path.")
    sys.exit()
```

```python
cv2.imshow("this is the first image", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

* **Requirement 2:** Print the image's height, width, and number of colour channels extracted from the image's .shape attribute.

```python
h, w, c = img.shape
print(f"Height: {h}px\nWidth: {w}px\nChannels: {c}")
```

* **Requirement 3:** Convert the image to grayscale using cv2.cvtColor() and display both the original colour image and the grayscale version in separate named windows.

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
cv2.imshow("Gray Image", gray)
cv2.imshow("Real Image(BGR)", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

* **Requirement 4:** Save the grayscale image to disk as grayscale_food.jpg using cv2.imwrite() and print a confirmation line showing the saved filename.

```python
cv2.imwrite("grayscale_food.jpg", gray)
print("Saved: grayscale_food.jpg")
```
