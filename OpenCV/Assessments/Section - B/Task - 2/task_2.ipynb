### Task 2: Menu Photo Resizer and Colour Channel Analyser

#### Build a program that resizes a restaurant menu photo to a standard model-input size and analyses its colour composition by channel.

* **Requirement 1:** Load a food image and resize it to exactly 256×256 pixels using cv2.resize() with the cv2.INTER_AREA interpolation flag.

```python
import cv2
import numpy as np
```

```python
img = cv2.imread('food_image.jpg')
resized = cv2.resize(img, (256, 256), interpolation=cv2.INTER_AREA)
```

```python
resized.shape
```

* **Requirement 2:** Split the resized image into its B, G, and R colour channels using cv2.split() and print the average pixel intensity of each channel rounded to two decimal places.

```python
b, g, r = cv2.split(resized)

avg_b = round(float(np.mean(b)), 2)
avg_g = round(float(np.mean(g)), 2)
avg_r = round(float(np.mean(r)), 2)
```

```python
print(f"Average Intensities:\nB: {avg_b}\nG: {avg_g}\nR: {avg_r}")
```

* **Requirement 3:** Display the resized image and all three individual channel images (as grayscale windows, one per channel) in separate named windows.

```python
cv2.imshow("Resized", resized)
cv2.imshow("Blue Channel", b)
cv2.imshow("Green Channel", g)
cv2.imshow("Red Channel", r)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

* **Requirement 4:** Print a one-line summary stating which colour channel has the highest average intensity and what that suggests about the dominant colour in the dish.

```python
channel_map = {'Blue': avg_b, 'Green': avg_g, 'Red': avg_r}
dominant = max(channel_map, key=channel_map.get)

print(f"The '{dominant}' channel has the highest average intensity, suggesting that warm red and orange tones (carrots, peppers, wooden board) \nare dominant in the food image.")
```
