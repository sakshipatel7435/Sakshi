### 1. SCENARIO
You are building a food delivery platform where restaurant partners upload menu item
photos. The product team wants to auto-categorise these images (e.g. biryani, pizza,
desserts) using Computer Vision instead of manual tagging.

* Question: Explain what Computer Vision is and how it differs from a simple rules-based image
filter. Why is a CV-based approach more suitable for categorising diverse food images than
writing fixed conditions based on colour thresholds or brightness alone?

### Ans:

* **CV Definition**: The field of AI that enables computers to derive meaningful information from visual inputs.  
* **Why Rule-Based Fails**: Simple color/brightness rules fail because lighting varies, dishes share colors (e.g., biryani, fried rice, curry are all yellow/orange), and presentation styles differ.  
* **CV Advantage**: Feature-based patterns and spatial relationships allow generalized categorization despite visual variations.

### 2. SCENARIO
You are working on a food quality inspection system that analyses photos of delivered
meals to check for freshness. The CV model was trained on small 128×128 grayscale
images, but restaurant partners upload large RGB photos up to 4000×3000 pixels

* **Preprocessing Needed**: Resizing to $128 \times 128$, color conversion, and intensity normalization.  
* **Why Resize**: Models require uniform input shapes to process batches efficiently.  
* **Grayscale Benefits**: Reduces memory footprint and computational cost by reducing data from 3 color channels to 1.  
* **When Grayscale Fails**: Freshness detection relies heavily on color cues (e.g., browning avocados or yellowing leafy greens).

### 3. SCENARIO
You are developing a packaging label scanner for a cloud kitchen. Delivery bags are
photographed by a ceiling-mounted camera, but bags sometimes arrive rotated at
random angles, and the region of interest (the nutrition label) is always in the
lower-right quadrant of the image.

* Question: Which OpenCV image transformation operations would you chain together to
normalise these images before passing them to an OCR engine? Justify the order in which you
would apply rotation, cropping, and flipping — and explain what goes wrong if you attempt to
crop before correcting the orientation.

### Ans:
**Correct Sequence:** Rotation $\rightarrow$ Cropping $\rightarrow$ Flipping.

**Justification:** 
* **1. Rotate:** Straighten the label so the bottom-right coordinate accurately captures the target region.
* **2.Crop:** Slice the bottom-right quadrant of the correctly oriented image.
* **3. Flip:** Fix any mirror reflection issues.
* **Failure Mode if Cropping First**: Cropping before rotation extracts the bottom-right region of an arbitrary/crooked view, missing the actual label.

### 4. SCENARIO
You are designing a burnt-food detection system for a dark kitchen. The system
captures images of dishes just before packing. A data scientist proposes using edge
detection to highlight char marks and overcooked crust regions rather than training a
full deep learning model.

* Question: Explain how the Canny edge detection algorithm works conceptually (gradient,
non-maximum suppression, hysteresis). Why is applying a Gaussian blur before Canny
important, and what trade-off exists between the low_threshold and high_threshold parameters
when tuning the detector to detect dark char marks without also highlighting every bright plate
rim?

### Ans:
**Conceptual Steps:**

* **1.Gradient Calculation:** Measures intensity changes across directions.

* **2.Non-Maximum Suppression:** Thins thick edges down to 1-pixel widths.

* **3.Hysteresis Thresholding:** Retains strong edges and weak edges connected to strong ones.

**Role of Gaussian Blur:** Reduces high-frequency noise so texture artifacts aren't mistaken for edges.

**Threshold Trade-off:** Lowering threshold detects dark char marks but picks up plate rims; raising it removes plate noise but misses subtle char marks.

### 5. SCENARIO
You are debugging a face detection module that verifies a delivery agent's identity
when they ring a customer's doorbell via a smart camera. The Haar cascade classifier is
raising false positives — it detects faces on picture frames and decorative objects
visible through the doorbell camera.


* Question: What causes false positives in Haar cascade face detection, and which parameters
— scaleFactor, minNeighbors, minSize — can you tune to reduce them? Explain the effect of
increasing minNeighbors and describe one specific scenario where setting it too high would
cause the module to fail in a real deployment.

### Ans:
**Causes:** Brightness patterns on non-face objects mimicking human facial features (eyes/nose bridge).
* **Parameter Fixes:** scaleFactor: Controls image pyramid scaling.
* **minNeighbors:** Higher values require more surrounding detections to confirm a face, reducing false positives.
* **minSize:** Ignores small artifacts.

**Failure Mode:** Setting minNeighbors too high (e.g., $>15$) skips real faces in poor lighting or slight angles.

### 6. SCENARIO
You are optimising a real-time video surveillance system for a dark kitchen that
monitors three food preparation stations simultaneously. The current implementation
reads every frame and writes it to disk, causing a 4-second lag and high storage costs
during peak hours.

* Question: Compare reading and processing every frame versus sampling frames at a fixed
interval in OpenCV. What strategy would you use to skip redundant frames without missing
important motion events, and how does the cap.set(cv2.CAP_PROP_POS_FRAMES, n) property
help implement controlled frame skipping?

### Ans:
* **Comparison:** Processing every frame burns excessive CPU/GPU resources and creates lag. Fixed-interval sampling lowers compute overhead while keeping real-time responsiveness.
* **Strategy:** Process every $N^{\text{th}}$ frame (e.g., every 5th frame).
* **Role of cap.set():** cap.set(cv2.CAP_PROP_POS_FRAMES, n) jumps directly to frame index n without decoding intermediate frames.
