##### Use an AI tool of your choice (ChatGPT, Claude, GitHub Copilot, etc.) to help you write a Python + OpenCV program that:

* Reads a food-item image from disk and applies a full processing pipeline: resize to 300×300, convert to grayscale, apply Gaussian blur (kernel 5×5), and run Canny edge detection.
* Counts and prints the total number of edge pixels detected in the processed image using
cv2.countNonZero().
* Displays the original colour image and the final edge map side-by-side in a single named
window using cv2.hconcat() or NumPy concatenation.
* Saves the edge map to disk as ai_edge_output.jpg and prints a confirmation message that
includes the total edge pixel count and the saved filename.


```python
import cv2
```

```python
img = cv2.imread('food_image.jpg')
```

```python
resized = cv2.resize(img, (300, 300))
gray = cv2.cvtColor(resized, cv2.COLOR_BGR2GRAY)
blurred = cv2.GaussianBlur(gray, (5, 5), 0)
edges = cv2.Canny(blurred, 100, 200)
```

```python
count = cv2.countNonZero(edges)
print(f"Total edge pixels: {count}")
```

```python
combined = cv2.hconcat([resized, edges])
```

```python
cv2.imwrite('ai_edge_output.jpg', combined)
print(f"Saved ai_edge_output.jpg with edge pixel count: {count}")
```

```python
cv2.imshow("Output", combined)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
