### TEST & DEBUG (WITHOUT AI)

```python
import cv2
```

```python
img = cv2.imread('food_image.jpg')
```

```python
# Fix 1: Added File Validation
if img is None:
    print(f"Error: Could not read image '{img}'")
```

```python
resized = cv2.resize(img, (300, 300), interpolation=cv2.INTER_AREA)
gray = cv2.cvtColor(resized, cv2.COLOR_BGR2GRAY)
blurred = cv2.GaussianBlur(gray, (5, 5), 0)
edges = cv2.Canny(blurred, 150, 300)
```

```python
count = cv2.countNonZero(edges)
```

```python
# Fix 2: Convert single-channel edge map to 3-channel (BGR) before concatenation
edges_bgr = cv2.cvtColor(edges, cv2.COLOR_GRAY2BGR)
combined = cv2.hconcat([resized, edges_bgr])
```

```python
cv2.imwrite('ai_edge_output.jpg', combined)
print(f"Successfully saved ai_edge_output.jpg | Total Edge Pixels: {count}")
```

```python
cv2.imshow("Original vs Edges", combined)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### Explanation of Corrections Made (3–4 Lines)
* The AI code failed during cv2.hconcat() because it attempted to concatenate a 3-channel color image (resized) with a 1-channel grayscale edge image (edges), resulting in an OpenCV dimension mismatch error. To fix this, I converted edges back into a 3-channel BGR image using cv2.cvtColor(edges, cv2.COLOR_GRAY2BGR) prior to concatenation. Additionally, I added file-existence validation to prevent program crashes if the target image is missing.
