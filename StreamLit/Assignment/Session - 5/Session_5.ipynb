### Tasks

* 1. Load a pre-trained IPL match outcome prediction model (saved as 'ipl_model.pkl') in a new Streamlit app and display a message confirming the model loaded successfully.

* 2. Create a Streamlit form that takes user input for a fantasy cricket app: team1 runs, team2 runs, and number of wickets left. On form submission, send the inputs to your loaded model and display the predicted winner.

* 3. Add error handling to your Streamlit prediction app so that if the model file is missing or corrupted, a user-friendly error message is shown instead of a crash.<br><br><em><strong>Hint:</strong> Use try-except around the model loading code and display errors using st.error().</em>

* 4. Modify your app to show the prediction result in a visually engaging way: if the predicted winner is team1, display a green success message; if team2, show a blue info message.<br><br><em><strong>Hint:</strong> Use st.success() and st.info() for different outcomes.</em>

* 5. Use ChatGPT or Copilot to help you write the code for taking multiple user inputs in a Streamlit form and sending them as a list to the model for prediction. Paste the generated code, then test and adapt it for your own app.
