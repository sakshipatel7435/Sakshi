### Tasks

* 1. Create a requirements.txt file for a Streamlit app that displays a list of trending movies using pandas and requests, making sure all necessary packages are included.

* 2. Push your Streamlit app folder (with requirements.txt and all code files) to a new public GitHub repository and share the repository link.

* 3. Set up a new Streamlit Cloud account, connect it to your GitHub repository containing your app, and deploy your app so it is accessible via a public URL.

* 4. Update your deployed app by changing the app title or adding a new text element, then push the changes to GitHub and verify that the live app reflects your update.<br><br><em><strong>Hint:</strong> Streamlit Cloud automatically redeploys when you push to the connected branch.</em>

* 5. Simulate a deployment error by removing a required package from requirements.txt, then redeploy your app and note the error message shown on Streamlit Cloud. Restore the package, redeploy, and confirm the app works.<br><br><em><strong>Hint:</strong> This helps you recognize and fix common deployment issues quickly.</em>
