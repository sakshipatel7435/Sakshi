### Tasks

* 1. Use st.line_chart to visualize your daily step count for the past 7 days in a Streamlit app, with days on the x-axis and steps on the y-axis.

* 2. Create a bar chart using st.bar_chart to compare the number of orders placed on Zomato, Swiggy, and Domino's over the last month. Use dummy data for each platform.

* 3. Build an area chart in Streamlit showing your Spotify listening time (in minutes) for each day of the week using st.area_chart.

* 4. Integrate Plotly in your Streamlit app to create an interactive pie chart showing the percentage of time you spend on Instagram, YouTube, and WhatsApp in a typical day.<br><br><em><strong>Hint:</strong> Use plotly.express.pie and st.plotly_chart to display the chart.</em>

* 5. Add a Streamlit multiselect widget that lets the user pick which columns (e.g., 'messages sent', 'photos shared', 'calls made') to visualize from a WhatsApp chat activity dataset, and display the selected data as a line chart.
